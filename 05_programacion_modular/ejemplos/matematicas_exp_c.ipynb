# 📦 Módulos en C: archivos `.h` y `.c`
*(comparación con la versión Python)*

---

## El modelo de módulos en C

Python tiene un concepto de módulo muy simple: **un archivo `.py`** es un módulo. En C, un "módulo" se compone de **dos archivos**:

* **`matematicas.h`** — la **cabecera**, que **declara** las funciones que ofrece el módulo.
* **`matematicas.c`** — la **implementación**, que define el código real de cada función.

Otros programas hacen `#include "matematicas.h"` y el compilador se encarga de juntar todo.

## El archivo de cabecera: `matematicas.h`

Solo **declara** las funciones (prototipos), sin implementarlas. Funciona como un **contrato**: anuncia *"este módulo ofrece estas funciones"*. Mira:

```c
/* matematicas.h */

#ifndef MATEMATICAS_H
#define MATEMATICAS_H

int mcd(int a, int b);
int es_primo(int n);
long factorial(int n);
double raiz_babilonica(double a);

#endif
```

Las líneas `#ifndef`/`#define`/`#endif` son una **guarda** para evitar que el archivo se incluya dos veces. Es una construcción muy estándar en C; tendrás que escribirla siempre en tus cabeceras.

## El archivo de implementación: `matematicas.c`

Aquí va el **código real** de las funciones. Incluye su propia cabecera (`#include "matematicas.h"`) para asegurarse de que las declaraciones coinciden con las implementaciones:

```c
/* matematicas.c */

#include "matematicas.h"

int mcd(int a, int b) {
    int resto;
    while (b != 0) {
        resto = a % b;
        a = b;
        b = resto;
    }
    return a;
}

/* ... resto de funciones ... */
```

Ambos archivos completos están en [matematicas.h](./matematicas.h) y [matematicas.c](./matematicas.c).

## El programa que usa el módulo: `usa_matematicas.c`

```c
/* usa_matematicas.c */

#include <stdio.h>
#include "matematicas.h"   // ← nuestra cabecera, con comillas (no <>)

int main(void) {
    printf("mcd(48, 18) = %d\n", mcd(48, 18));
    printf("5! = %ld\n", factorial(5));
    return 0;
}
```

> 💡 **Detalle importante**: las cabeceras del sistema van con corchetes `<>` (`<stdio.h>`), las nuestras con comillas `""` (`"matematicas.h"`). Eso le dice al compilador dónde buscarlas.

## Compilando todo junto

Para compilar un programa que usa un módulo en C, hay que pasar **todos los archivos `.c`** al compilador:

```bash
gcc -o programa usa_matematicas.c matematicas.c -lm
```

El compilador combina los dos archivos en un único ejecutable `programa`. Si tu proyecto crece a 20 archivos, esto se vuelve tedioso, por eso se usan herramientas como **`make`** o **`CMake`** para automatizarlo. Pero para proyectos pequeños como los nuestros, llamar a `gcc` directamente es suficiente.

## Comparativa Python vs C

| Aspecto | Python | C |
| :--- | :--- | :--- |
| Qué es un módulo | Un archivo `.py` | Dos archivos: `.h` (declara) + `.c` (define) |
| Importar | `import matematicas` | `#include "matematicas.h"` |
| Compilar | (no hace falta, Python es interpretado) | Hay que incluir todos los `.c` en el comando |
| Evitar incluir dos veces | (automático) | `#ifndef`/`#define`/`#endif` |
| Ámbito del módulo | El archivo entero | Funciones `static` para limitar al `.c` |

Cuando se pone `static` antes de una función en C, le estás diciendo al compilador: "*Esta función **solo existe dentro de este archivo `.c`**. Nadie fuera de este archivo puede verla ni llamarla*". Si no usas `static`, corres el riesgo de que dos archivos `.c` diferentes tengan una función llamada igual (por ejemplo, `auxiliar()`), lo que provocará un error de colisión al compilar ("multiple definition").  

## 🎯 Lo que debes recordar

* ✅ En C, **un módulo = cabecera (`.h`) + implementación (`.c`)**.
* ✅ La cabecera **declara** (qué ofrece el módulo); la implementación **define** (cómo lo hace).
* ✅ La **guarda con `#ifndef`/`#define`** evita inclusiones duplicadas.
* ✅ Para compilar, **todos los `.c`** se pasan al comando `gcc`.
* ✅ Es **más laborioso que en Python**, pero el principio (separar declaración de implementación, organizar el código en piezas) **es el mismo**.
