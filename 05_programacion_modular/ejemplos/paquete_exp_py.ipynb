## Cuando un módulo se queda pequeño

En el ejemplo anterior reunimos las funciones matemáticas en un único archivo `matematicas.py`. Eso es suficiente para 5 o 10 funciones. Pero, ¿qué pasa cuando tienes **80 funciones** distribuidas en geometría, aritmética, estadística, álgebra lineal...?

Un solo archivo se vuelve **inmanejable**. La solución es **organizar las funciones en carpetas temáticas**: es lo que llamamos un **paquete** con **submódulos**.

## La estructura del paquete

Vamos a crear un paquete `paq_matematicas` con dos áreas:

```text
paq_matematicas/
├── __init__.py             ← marca esta carpeta como paquete
├── aritmetica/
│   ├── __init__.py         ← marca esta carpeta como subpaquete
│   └── enteros.py          ← submódulo: MCD, MCM, primos
└── geometria/
    ├── __init__.py
    └── plana.py            ← submódulo: áreas, perímetros
```

**Tres ideas clave**:

* Una **carpeta con un `__init__.py`** dentro es un **paquete**.
* Una carpeta dentro de un paquete con su propio `__init__.py` es un **subpaquete**.
* El nombre del paquete es el de la carpeta. Para acceder a un submódulo: `paquete.submódulo`.

## El archivo `__init__.py`

Es un archivo especial cuya presencia le dice a Python *"esta carpeta es un paquete"*. **Puede estar vacío** (muchas veces lo está) o contener una **docstring** describiendo el paquete. En este ejemplo, el contenido es:

```python
"""
paq_matematicas — Paquete de utilidades matemáticas con submódulos.
"""
```

Y cada subpaquete tiene también el suyo:

```python
# paq_matematicas/aritmetica/__init__.py
"""Submódulos: enteros."""
```

## El contenido de un submódulo

El archivo `paq_matematicas/aritmetica/enteros.py` contiene funciones normales y corrientes:

```python
def mcd(a: int, b: int) -> int:
    """Algoritmo de Euclides."""
    while b != 0:
        a, b = b, a % b
    return a

def es_primo(n: int) -> bool:
    if n < 2: return False
    if n == 2: return True
    if n % 2 == 0: return False
    i = 3
    while i * i <= n:
        if n % i == 0: return False
        i = i + 2
    return True
```

Ya has visto varias implementaciones de estas funciones, como puedes observar un mismo problema puede tener varias soluciones. Una más para `es_primo` (en un estilo mucho menos **"pitónico"**):

```python
def es_primo(n):
    if n < 2: primo=False
    elif n == 2: primo=True
    elif n % 2 == 0: primo=False
    else:
        primo=True
        i = 3
        while (i * i <= n) and primo:
            if n % i == 0: primo=False
            i = i + 2
    return primo
```

## Importando desde el paquete

Veamos las distintas formas de importar. Imagina que en la misma carpeta donde está la carpeta `paq_matematicas/` tienes un programa `mi_programa.py`:

````python
# Forma 1: importar el submódulo completo con la ruta completa
import paq_matematicas.aritmetica.enteros

print(paq_matematicas.aritmetica.enteros.mcd(48, 18))
````

La forma anterior funciona, pero es **pesada de escribir**. Por eso lo más habitual es usar un alias:

````python
# Forma 2: importar con alias (¡la más útil!)
import paq_matematicas.aritmetica.enteros as aritm
import paq_matematicas.geometria.plana as geo

print(aritm.mcd(48, 18))
print(geo.area_circulo(5))
````

````python
# Forma 3: importar funciones concretas
from paq_matematicas.aritmetica.enteros import mcd, es_primo

print(mcd(48, 18))
print(es_primo(7))
````

## ¿Te suena? NumPy hace exactamente lo mismo

Lo que acabas de crear es **estructuralmente idéntico** a lo que hacen las grandes bibliotecas científicas. Por ejemplo, cuando escribes:

```python
import numpy as np
np.linalg.solve(A, b)        # álgebra lineal
np.random.normal(0, 1)        # números aleatorios
np.fft.fft(senal)             # transformada de Fourier
```

estás llamando a funciones de los **submódulos** `linalg`, `random` y `fft` del paquete `numpy`. **Misma idea, distinta escala.**

## 🎯 Conceptos del Tema 5 que has practicado

* ✅ **Paquete** = carpeta con `__init__.py`.
* ✅ **Subpaquete** = carpeta dentro de un paquete, también con `__init__.py`.
* ✅ **Submódulo** = archivo `.py` dentro de un (sub)paquete.
* ✅ **Importación con ruta completa** y **alias**.
* ✅ Reconocer la estructura de paquetes como **el patrón estándar** del ecosistema científico Python.
