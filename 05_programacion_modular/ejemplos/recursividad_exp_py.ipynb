# 🪞 Recursividad: cuando una función se llama a sí misma
*(versión explicada paso a paso)*

---

## Una idea aparentemente paradójica

Una función **recursiva** es una función que **se llama a sí misma**. La primera vez que se ve, uno piensa: *"¿pero eso no se quedará en un bucle infinito?"*

La clave está en que **cada llamada se hace con un argumento más simple** que la anterior, hasta llegar a un **caso base** que se resuelve directamente sin más llamadas.

## La fórmula universal

Toda función recursiva tiene **dos partes**:

1. **Caso base**: un valor para el que conocemos la respuesta directamente.
2. **Caso recursivo**: cómo expresar el problema en términos de uno más pequeño.

Por ejemplo, el factorial $n!$ se puede definir así:

$$
n! = \begin{cases}
    1 & \text{si } n = 0 \text{ o } n = 1 & \text{(caso base)} \\
    n \cdot (n-1)! & \text{si } n > 1 & \text{(caso recursivo)}
\end{cases}
$$

Y la traducción a Python es **literal**:

```python
def factorial_recursivo(n: int) -> int:
    if n <= 1:
        return 1                              # ← caso base
    return n * factorial_recursivo(n - 1)   # ← caso recursivo

print(f'5! = {factorial_recursivo(5)}')
print(f'10! = {factorial_recursivo(10)}')
```

> 💡 **Cómo funciona internamente**: para calcular `factorial_recursivo(5)`, Python:
>
> 1. Necesita `5 * factorial_recursivo(4)`. Llama a `factorial_recursivo(4)`.
> 2. Para ese cálculo, necesita `4 * factorial_recursivo(3)`. Llama...
> 3. ... y así hasta `factorial_recursivo(1)` que devuelve 1 directamente.
> 4. Entonces los valores **van subiendo**: 1 → 2 → 6 → 24 → 120.
>
> Python guarda cada llamada pendiente en una estructura llamada **pila de llamadas** (*call stack*).

## Comparación con la versión iterativa

La misma cuenta se puede hacer con un bucle `for` (lo que sabíamos del Tema 4):

```python
def factorial_iterativo(n: int) -> int:
    resultado = 1
    for i in range(2, n + 1):
        resultado = resultado * i
    return resultado

print(f'5! = {factorial_iterativo(5)}')
print(f'10! = {factorial_iterativo(10)}')
```

**Ambas versiones devuelven el mismo resultado.** Para el factorial, las dos son igual de eficientes. Pero para otros problemas, la diferencia puede ser **dramática**, como veremos en el ejemplo siguiente.

## El caso Fibonacci: cuando la recursividad es muy elegante… pero muy mala idea

La sucesión de Fibonacci se define recursivamente como:

$$F_n = F_{n-1} + F_{n-2}, \quad F_0 = 0, \quad F_1 = 1$$

Y la traducción a Python es preciosa:

```python
def fibonacci_recursivo(n: int) -> int:
    if n <= 1:
        return n
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)

print(f'fib(10) = {fibonacci_recursivo(10)}')
```

> ⚠️ **Pero hay un problema gravísimo**: para calcular `fibonacci_recursivo(n)`, esta función llama dos veces a sí misma. Y cada una de esas llamadas llama otras dos veces. **Acaba haciendo cantidades astronómicas de llamadas repetidas**. Por ejemplo, para calcular `fibonacci_recursivo(35)` el ordenador tarda **más de 1 segundo**. Para `fibonacci_recursivo(50)` tardaría varios minutos.

Y aquí viene lo dramático:

```python
def fibonacci_iterativo(n: int) -> int:
    if n <= 1:
        return n
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

import time

n = 35
t0 = time.time()
fibonacci_recursivo(n)
t_rec = time.time() - t0

t0 = time.time()
fibonacci_iterativo(n)
t_iter = time.time() - t0

print(f'Recursivo: {t_rec:.3f} s')
print(f'Iterativo: {t_iter*1e6:.1f} μs')
print(f'\nLa versión iterativa es ~{t_rec/t_iter:.0f} veces más rápida.')
```

**¡Factor 100 000!** Misma operación. Misma fórmula matemática. Distinta forma de implementarla.

## Conclusión: ¿cuándo usar recursividad?

### ✅ Cuándo SÍ usarla

* Cuando el problema tiene **estructura recursiva natural**: factoriales, recorridos de árboles, divide y vencerás (mergesort, quicksort), búsquedas en estructuras anidadas...
* Cuando la versión recursiva es **notablemente más clara** que la iterativa.

### ❌ Cuándo NO usarla

* Cuando una iteración hace lo mismo **más rápido y con menos memoria**.
* Cuando se solapan recálculos masivamente (como en Fibonacci) y no usas **memoización**. La memoización es una técnica de programación que consiste en almacenar en memoria el resultado de una función para reutilizarlo después. Así, si la función vuelve a ejecutarse con los mismos parámetros, el programa devuelve el resultado guardado en lugar de recalcularlo desde cero.  
* Cuando la profundidad de recursión pueda superar el **límite por defecto de Python** (≈ 1000).
