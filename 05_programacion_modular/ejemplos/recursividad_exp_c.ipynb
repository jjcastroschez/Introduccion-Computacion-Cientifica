# 🪞 Recursividad en C
*(comparación con la versión Python)*

---

La recursividad funciona **exactamente igual** en C que en Python: una función se llama a sí misma. La sintaxis cambia, pero las ideas (caso base, caso recursivo, pila de llamadas) son **idénticas**.

## El factorial en C

```c
long factorial_recursivo(int n) {
    if (n <= 1) return 1;             // caso base
    return n * factorial_recursivo(n - 1);   // caso recursivo
}
```

Y Python (idéntico):

```python
def factorial_recursivo(n: int) -> int:
    if n <= 1: return 1
    return n * factorial_recursivo(n - 1)
```

> 💡 **¿Por qué `long` en lugar de `int` para el resultado?** Porque el factorial crece muy rápido: `20!` ya se sale del rango de un `int` de 32 bits. En Python esto no es un problema porque los enteros tienen precisión arbitraria; en C, **debes elegir el tipo conscientemente**.

## Fibonacci recursivo: la trampa de la ineficiencia

Igual que en Python, la versión recursiva ingenua de Fibonacci es desastrosamente lenta:

```c
long fibonacci_recursivo(int n) {
    if (n <= 1) return n;
    return fibonacci_recursivo(n-1) + fibonacci_recursivo(n-2);
}
```

Si compilas y ejecutas `recursividad.c` con `n=35`, verás algo así:

```text
fibonacci_recursivo(35) = 9227465 (0.083 s)
fibonacci_iterativo(35) = 9227465 (0.000001 s)
```

**C es más rápido que Python** (en Python la recursiva con `n=35` tarda 1 segundo, en C 0.08 s) pero **la diferencia entre recursiva e iterativa sigue siendo abismal**: factor 100 000.

> 🎯 **Conclusión universal**: la elección entre recursión e iteración es una decisión **algorítmica**, no del lenguaje. Si el algoritmo recursivo es exponencial, lo será en cualquier lenguaje.

## Diferencias prácticas con Python

### 1. Pila de llamadas (*call stack*)

Ambos lenguajes mantienen una pila de llamadas, pero los **límites son distintos**:

* **Python**: ≈ 1000 llamadas anidadas por defecto. Si lo superas, lanza `RecursionError`.
* **C**: depende del sistema operativo (típicamente ≈ 1 MB de pila, suficiente para decenas de miles de llamadas), pero al superarla provoca un **stack overflow** que ¡suele caer en *segmentation fault*!

Python te avisa con un error claro; C puede no avisarte. Por eso, **en C debes ser aún más cuidadoso** con la profundidad de recursión.

### 2. Tipos de retorno

En Python, una función recursiva como `factorial` devuelve un **entero de precisión arbitraria**: puedes calcular `factorial(100)` y obtener un número de 158 cifras sin problemas.

En C, debes elegir un tipo (`int`, `long`, `long long`) y **te quedas con su rango**. Para factoriales grandes necesitarías bibliotecas de aritmética de precisión arbitraria (como GMP).
