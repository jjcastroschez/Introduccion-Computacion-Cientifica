## La sintaxis cambia, la idea es la misma

La definición de funciones es probablemente la construcción que **más cambia** al pasar de Python a C, pero los **conceptos** son idénticos: parámetros, valor de retorno y reutilización.

## El código en C

```c
#include <stdio.h>
#include <math.h>

/* Prototipos de las funciones — declaración "adelantada" */
double area_circulo(double radio);
double hipotenusa(double cateto1, double cateto2);
int es_par(int n);

int main(void) {
    printf("Área(5) = %.4f\n", area_circulo(5));
    printf("Hip(3,4) = %.4f\n", hipotenusa(3, 4));
    printf("7 es par? %d\n", es_par(7));
    return 0;
}

/* Definiciones */
double area_circulo(double radio) {
    return M_PI * radio * radio;
}

double hipotenusa(double cateto1, double cateto2) {
    return sqrt(cateto1 * cateto1 + cateto2 * cateto2);
}

int es_par(int n) {
    return n % 2 == 0;
}
```

El archivo completo está en [calculadora.c](./calculadora.c).

## Diferencias clave con Python

### 1. Tipos explícitos en parámetros y retorno

**Python**:
```python
def area_circulo(radio: float) -> float:
```
Los `: float` y `-> float` son **opcionales** (type hints, solo orientativos).

**C**:
```c
double area_circulo(double radio)
```
Los tipos son **obligatorios**. El compilador los necesita para reservar memoria, generar instrucciones y comprobar errores. Si pasas un `int` donde se espera un `double`, el compilador lo **convertirá** (no fallará), pero si el tipo no encaja se queja explícitamente.

### 2. La función `main` es obligatoria

En Python, el código fuera de funciones se ejecuta directamente. En C, **toda la ejecución empieza** obligatoriamente en una función llamada `main` que devuelve un entero (`0` significa "OK"). No hay nada parecido al `if __name__ == "__main__":` de Python — en C, **siempre** hay un único punto de entrada.

### 3. Prototipos

En C, **antes de usar una función**, el compilador necesita conocer su "firma" (tipos de parámetros y retorno). Por eso es habitual escribir los **prototipos** al principio del archivo (líneas como `double area_circulo(double radio);`) y las definiciones completas más abajo. Es parecido a una **declaración adelantada**. En Python esto no es tan estricto: ya que se lee todo el archivo antes de ejecutar. Por lo que podríamos definir una función que usa unas funciones declaradas posteriormente. La única regla es que cuando se comienza a ejecutar código lo que necesita ya esté declarado.   

### 4. Cabeceras y bibliotecas

```c
#include <stdio.h>   // para printf, scanf
#include <math.h>    // para sqrt, M_PI
```

Y al compilar, hay que **enlazar la biblioteca matemática** con `-lm`:

```bash
gcc -o calculadora calculadora.c -lm
```

Si olvidas el `-lm`, te dará `undefined reference to 'sqrt'`.

## Compilar y ejecutar

```bash
gcc -o calculadora calculadora.c -lm
./calculadora
```

## 🎯 Lo que debes recordar

* ✅ La **idea de función** (parámetros + valor de retorno) es **idéntica** en ambos lenguajes.
* ✅ En C, los **tipos son obligatorios** y se declaran tanto para los parámetros como para el valor de retorno.
* ✅ En C, la ejecución **siempre empieza** en `main()`.
* ✅ Para usar `sqrt`, `M_PI`, etc., necesitas `#include <math.h>` Y compilar con `-lm`.
