# 🛡️ Validación de entrada como función — versión en C
*(comparación con la versión Python)*

---

Este ejemplo en C es interesante porque **C no tiene `try-except`**. Aun así, podemos encapsular la validación en una función reutilizable, usando las técnicas que vimos en el [Tema 4](../../04_control_flujo_ejecucion/ejemplos/T4_Ejem_ICC.md): comprobar el valor de retorno de `scanf` y vaciar el buffer.

## El código en C

```c
#include <stdio.h>

/*
 * Pide un entero al usuario hasta que sea válido y esté en [minimo, maximo].
 */
int pedir_entero(const char *mensaje, int minimo, int maximo) {
    int n;
    int caracter;

    while (1) {
        printf("%s", mensaje);
        if (scanf("%d", &n) != 1) {
            printf("  ❌ Eso no es un entero válido.\n");
            while ((caracter = getchar()) != '\n' && caracter != EOF);
            continue;
        }
        if (minimo <= n && n <= maximo) {
            return n;
        }
        printf("  ❌ %d está fuera del rango [%d, %d].\n", n, minimo, maximo);
    }
}
```

El archivo completo (con `pedir_real_positivo` también) está en [validacion_entrada.c](./validacion_entrada.c).

👀 Observa el uso de `while (1)` es el equivalente al `while True` de Python. En C no existe la constante `True` por defecto (se puede definir con `#include <stdbool.h>`, pero tradicionalmente se usa `1` para *cierto* y `0` para *falso*). La condición `while (1)` es un bucle infinito que se rompe con `return` o `break`.

Aquí, en C, esta construcción es menos legible todavía, se violan los principios de la programación estructurada, mejor con la **variable de control** o **bandera** (también llamada flag en inglés).

## El código en C con variable 

```c
#include <stdio.h>
#include <stdbool.h> // Librería obligatoria para usar bool

/*
 * Pide un entero al usuario hasta que sea válido y esté en [minimo, maximo].
 */
int pedir_entero(const char *mensaje, int minimo, int maximo) {
    int n=0;
    int caracter;
    bool entrada_valida;

    entrada_valida=false;
    while (!entrada_valida) {
        printf("%s", mensaje);
        if (scanf("%d", &n) != 1) {
            printf("  ❌ Eso no es un entero válido.\n");
            while ((caracter = getchar()) != '\n' && caracter != EOF);
        }
        else if (minimo <= n && n <= maximo) {
                 entrada_valida=true;
             }
             else { 
                  printf("  ❌ %d está fuera del rango [%d, %d].\n", n, minimo, maximo);
             }
    }
    return n;
}
```



## Diferencias clave con Python

### 1. Cadenas como parámetro: `const char *`

Python:
```python
def pedir_entero(mensaje: str, minimo: int, maximo: int) -> int:
```

C:
```c
int pedir_entero(const char *mensaje, int minimo, int maximo)
```

En C, una cadena de texto no es un tipo primitivo: es un **puntero a un array de caracteres**. El `const char *` significa *"una dirección de memoria que apunta a caracteres que **no vamos a modificar**"*. Si quitas el `const`, también compila, pero indicarlo es buena práctica.

Los punteros es un tema complejo de C, intentaremos profundizar algo más adelante en el curso. De momento te basta con saber que **`const char *` es la forma C de pasar una cadena a una función**.

### 2. No hay `try-except`, hay valor de retorno de `scanf`

**Python**:
```python
try:
    n = int(entrada)
except ValueError:
    # gestionar el error
```

**C**:
```c
if (scanf("%d", &n) != 1) {
    // gestionar el error
}
```

`scanf` devuelve **el número de elementos que ha conseguido leer**. Si pides un `%d` y el usuario teclea letras, `scanf` devuelve `0`. Comprobamos ese retorno para detectar el error.

### 3. Hay que vaciar el buffer

Cuando `scanf` falla porque no encuentra un entero, los caracteres mal formados **se quedan en el buffer** de entrada. Si no los vaciamos, el siguiente `scanf` fallará otra vez, indefinidamente.

```c
while ((caracter = getchar()) != '\n' && caracter != EOF);
```

Este pequeño bucle **lee y descarta** todos los caracteres hasta encontrar un salto de línea o el fin del archivo. **Es un patrón que debes conocer**: aparece en casi todos los programas C que leen del teclado.

## Comparativa lado a lado

| Aspecto | Python | C |
| :--- | :--- | :--- |
| Cadena como parámetro | `mensaje: str` | `const char *mensaje` |
| Manejo de errores | `try / except ValueError` | `if (scanf(...) != 1)` |
| Vaciar buffer tras error | (automático) | `while (getchar() != '\n');` |
| Bucle hasta éxito | `while True` | `while (1)` |
| Salir del bucle | `return n` | `return n` |

## 🎯 Lo que debes recordar

* ✅ La idea de **encapsular validación en una función reutilizable** es universal y aplica también a C.
* ✅ En C, las cadenas son punteros: `const char *` para parámetros de tipo texto.
* ✅ Sin `try-except`, comprobamos el valor de retorno de `scanf` y vaciamos el buffer manualmente.
