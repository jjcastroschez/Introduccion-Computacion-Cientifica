## Lo que aporta el Tema 5

Encapsulamos ese patrón en una **función reutilizable**. Una vez bien escrita y probada, **la usamos siempre** con un simple cambio de argumentos.

## Paso 1: Definir la función de validación

La idea es que la función reciba **lo que cambia entre una llamada y otra**: el mensaje, el rango mínimo y el rango máximo. El resto (bucle, try-except, mensajes de error) queda encapsulado dentro:

````python
def pedir_entero(mensaje: str, minimo: int, maximo: int) -> int:
    """
    Pide al usuario un entero hasta que sea válido y esté en [minimo, maximo].
    """
    while True:
        entrada = input(mensaje)
        try:
            n = int(entrada)
        except ValueError:
            print(f'  ❌ "{entrada}" no es un entero válido.')
            continue
        if minimo <= n <= maximo:
            return n
        print(f'  ❌ {n} está fuera del rango [{minimo}, {maximo}].')
````

> 💡 **Observa el `return n` en medio del bucle**: al ejecutar `return`, la función termina inmediatamente y devuelve el valor. **No es necesario poner `break` adicional** — el `return` ya rompe el bucle. Observa también el uso del `continue` cuando se lanza la excepción, implica ir a la siguiente ejecución del `while True`.

Este código es más "Pythonic", pero probablemente menos claro, en estos primeros pasos de tu aprendizaje, que este otro:

````python
def pedir_entero(mensaje: str, minimo: int, maximo: int) -> int:
    """
    Pide al usuario un entero hasta que sea válido y esté en [minimo, maximo].
    """
    entrada_valida = False
    while not entrada_valida:
        entrada = input(mensaje)
        try:
            n = int(entrada)
        except ValueError:
            print(f'  ❌ "{entrada}" no es un entero válido.')
        else:
            if minimo <= n <= maximo:
                entrada_valida = True
            else:
                print(f'  ❌ {n} está fuera del rango [{minimo}, {maximo}].')
    return n
````

👀 Observa que este último código tiene un punto de entrada (al comienzo) y uno de salida (al final, con el return). El código anterior lo tiene más escondido 😜 lo cual puede hacer tu código más difícil de entender 🤯. 

## Paso 2: Usar la función las veces que haga falta

Ahora pedir varios datos es **trivial**:

````python
# Por brevedad, simulemos respuestas válidas
# En un programa real, estos input() pedirían al usuario
import io, sys
sys.stdin = io.StringIO('hola\n25\n200\n42\n') 

edad = pedir_entero('Edad: ', 16, 100)
puntos = pedir_entero('Puntos: ', 0, 100)

print(f'\nEdad capturada: {edad}')
print(f'Puntos capturados: {puntos}')
````

> 🎯 **Lo importante**: con dos líneas `edad = pedir_entero(...)` y `puntos = pedir_entero(...)` hemos obtenido lo mismo que antes requería **20+ líneas duplicadas**. Y si mañana decidimos mejorar el mensaje de error, lo cambiamos **una sola vez** en la función y todas las llamadas heredan la mejora automáticamente.

## Paso 3: Otra función de validación para reales positivos

Aplicamos la misma idea para reales que deben ser positivos:

````python
def pedir_real_positivo(mensaje: str) -> float:
    """Pide un real estrictamente positivo hasta que sea válido."""
    entrada_valida = False
    while not entrada_valida:
        entrada = input(mensaje)
        try:
            x = float(entrada)
        except ValueError:
            print(f'  ❌ "{entrada}" no es un número real válido.')
        else:
            if x > 0:
                entrada_valida = True
            else:
                print(f'  ❌ {x} no es positivo.')
    return x    
````

## Paso 4: Programa principal corto, claro y robusto

Veamos cómo queda un programa que calcule el IMC del usuario. Compáralo mentalmente con la cantidad de código que habría hecho falta sin funciones:

````python
import io, sys
sys.stdin = io.StringIO('25\n1.78\n75\n')

edad = pedir_entero('Tu edad (entre 16 y 100): ', 16, 100)
altura = pedir_real_positivo('Tu altura en metros: ')
peso = pedir_real_positivo('Tu peso en kilos: ')

imc = peso / (altura ** 2)
print(f'\nTu IMC es {imc:.2f}')
````

## 🎯 Conceptos del Tema 5 que has practicado

* ✅ **Encapsulación**: meter dentro de una función todo lo que se repite.
* ✅ **Parámetros como puntos de variación**: lo que cambia entre llamadas se pasa como argumento.
* ✅ **`return` dentro de un bucle**: termina la función inmediatamente, sin necesidad de `break`.
* ✅ **Reutilización**: ahora puedes usar `pedir_entero` y `pedir_real_positivo` en TODOS tus programas.
