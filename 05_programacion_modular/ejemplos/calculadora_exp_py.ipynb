## El primer paso: definir funciones propias

Hasta el [Tema 4](../../04_control_flujo_ejecucion/README.md), todo nuestro código vivía dentro de un único bloque. Cuando necesitábamos calcular algo dos veces (un MCD para dos pares de números, una raíz babilónica para dos valores...), **copiábamos y pegábamos** el código.

A partir del Tema 5 dejamos de hacer eso. **Definimos funciones** que encapsulan los cálculos y los podemos invocar tantas veces como queramos, desde cualquier punto del programa.

## Estructura general de una función en Python

```python
def nombre_funcion(parametros) -> tipo_devuelto:
    """Docstring: qué hace, qué espera, qué devuelve."""
    # cuerpo de la función
    return resultado
```

* `def` es la palabra clave que **introduce** la definición.
* `parametros` es la lista de entradas (separadas por comas).
* `-> tipo_devuelto` es opcional (type hint) pero muy útil para legibilidad.
* `return` devuelve el resultado. Si no escribes `return`, la función devuelve `None`.

Vamos a empezar con la función más simple posible: el **área de un círculo**.

````python
import math

def area_circulo(radio: float) -> float:
    """Devuelve el área de un círculo de radio dado."""
    return math.pi * radio ** 2
````

👀 Observa como el parámetro *radio* de la función `area_circulo`, tiene asociado el tipo *float* (`radio: float`), esto es también type hint. Recuerda: no es necesario, solo es usado para aumentar la legibilidad del código.   

Una vez que está definida, ya podemos **usarla** las veces que queramos:

````python
print(f'Área r=1: {area_circulo(1):.4f}')
print(f'Área r=5: {area_circulo(5):.4f}')
print(f'Área r=10: {area_circulo(10):.4f}')
````

> 💡 **Antes** habrías tenido que escribir `math.pi * 1**2`, `math.pi * 5**2`, `math.pi * 10**2`. **Ahora** delegas el cálculo en la función y solo cambias el argumento. Si mañana decides usar una fórmula distinta (por ejemplo, una corrección por curvatura), **modificas la función una sola vez** y todas las llamadas siguen funcionando.

## Funciones con varios parámetros

La hipotenusa de Pitágoras necesita dos parámetros, los dos catetos:

````python
def hipotenusa(cateto1: float, cateto2: float) -> float:
    """Hipotenusa de un triángulo rectángulo."""
    return math.sqrt(cateto1 ** 2 + cateto2 ** 2)

print(f'h(3, 4) = {hipotenusa(3, 4)}')
print(f'h(5, 12) = {hipotenusa(5, 12)}')
````

> 💡 **Observa la separación de responsabilidades**: la función `hipotenusa` no sabe (ni le importa) de dónde vienen los catetos. Pueden ser entradas del usuario, valores fijos, o resultados de otra función. Eso es lo que hace que sea **reutilizable**.

## Funciones que devuelven un booleano

Una convención muy útil: las funciones que devuelven `True` o `False` (booleanos) se nombran con verbos como `es_`, `tiene_`, `hay_`, `esta_`. Hacen el código mucho más legible:

````python
def es_par(n: int) -> bool:
    """Devuelve True si n es par, False en caso contrario."""
    return n % 2 == 0

for num in range(1, 6):
    if es_par(num):
        print(f'{num} es par')
    else:
        print(f'{num} es impar')
````

> 🎯 **Léelo en voz alta**: *"si es par num, imprime que num es par"*. **El código se lee como inglés natural.** Esto es lo que se entiende por código limpio. Compáralo con `if num % 2 == 0:` — funciona, pero es más críptico.

## Estructura de un programa modular: la función `main()`

Una buena práctica es **encapsular el programa principal dentro de una función `main()`** y protegerla con la guarda `if __name__ == "__main__":`. Esto tiene dos ventajas:

1. **Reutilizable**: si otro programa hace un `import` de nuestro `.py`, las funciones de él quedan disponibles, pero el `main()` no se ejecuta automáticamente.
2. **Organizado**: queda muy claro dónde empieza el flujo principal.

```python
def main():
    # ... lógica principal ...

if __name__ == "__main__":
    main()
```

Verás esta construcción en **todos** los programas Python profesionales. Esta misma estructura aparece en el script [calculadora.py](./calculadora.py).
