## ¿Qué es un módulo?

Un **módulo** en Python es, sencillamente, **un archivo `.py`**. Eso es todo. Cuando escribes `import math`, Python busca un archivo `math.py` (o equivalente compilado) y ejecuta su contenido para ponerlo a tu disposición.

Lo bueno es que **tú también puedes crear tus propios módulos**. Y eso es exactamente lo que vamos a hacer ahora: reunir todas las funciones matemáticas que escribimos en el Tema 4 (Euclides, raíz babilónica, criba…) en un solo archivo `matematicas.py` que podremos importar desde cualquier programa.

## Por qué esto es un salto cualitativo

Recuerda cómo eran los programas del Tema 4: cada ejercicio era un script independiente que **no podía comunicarse con otro**. Si querías reutilizar el algoritmo de Euclides para otro programa, tenías que **copiar y pegar** el código.

Con módulos, **eso se acaba**. Escribes la función una vez, la pones en un módulo, y la importas donde la necesites.

## El archivo `matematicas.py`

Aquí tienes el módulo completo que vamos a crear. Cópialo a un archivo llamado `matematicas.py` (la carpeta donde lo guardes es la misma donde luego escribirás los programas que lo usen).

```python
"""
matematicas.py — Módulo con funciones matemáticas útiles.
"""

import math

def mcd(a: int, b: int) -> int:
    """Máximo Común Divisor por el algoritmo de Euclides."""
    while b != 0:
        a, b = b, a % b
    return a

def es_primo(n: int) -> bool:
    """Devuelve True si n es primo, False en caso contrario."""
    if n < 2: return False
    if n == 2: return True
    if n % 2 == 0: return False
    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i = i + 2
    return True


def raiz_babilonica(a: float, tolerancia: float = 1e-12) -> float:
    """Raíz cuadrada por el método babilónico (Herón)."""
    if a < 0:
        raise ValueError('No existe raíz real de un negativo.')
    if a == 0: return 0.0
    x = a / 2
    for _ in range(100):
        x_nuevo = 0.5 * (x + a / x)
        if abs(x_nuevo - x) < tolerancia: return x_nuevo
        x = x_nuevo
    return x
```

El archivo completo, con más funciones, está en [matematicas.py](./matematicas.py). Ahí, en ese archivo he puesto las versiones menos "pythonic", recuerda "siguen más las recomendaciones de la programación estructurada". 

Antes de avanzar, 👀 observa el uso de `raise ValueError('No existe raíz real de un negativo.')`. La instrucción `raise` en Python sirve para *lanzar o generar una excepción de manera intencional*. Y luego se usaría donde se invoque la función:

```python
    try:
        # Esto lanzará el ValueError
        resultado = raiz_babilonica(-5)
    except ValueError as e:
        # Aquí capturas el error y manejas la situación de forma limpia
        print(f"Lanzamiento de mensaje esperado: {e}")
    else:
        print(f"raiz_babilonica(-5) = {resultado}")
```

## La guarda `if __name__ == "__main__":`

Al final del módulo encontrarás esta construcción mágica:

```python
if __name__ == "__main__":
    print('Tests rápidos del módulo')
    print(f'  mcd(48, 18) = {mcd(48, 18)}')
    ...
```

**¿Qué hace?** Cuando Python ejecuta un archivo, define automáticamente una variable interna `__name__`. Esta variable vale:
* `"__main__"` si el archivo se ejecuta **directamente** (`python matematicas.py`).
* `"matematicas"` (el nombre del módulo) si se **importa** desde otro archivo.

Por lo tanto, el código dentro del `if` solo se ejecuta cuando lanzamos `matematicas.py` desde la terminal —usándolo como un programa— pero **no** cuando otro programa lo importa. Es la forma profesional de hacer que un archivo Python sea simultáneamente:

1. Una **biblioteca** importable.
2. Un **programa ejecutable** (con sus propios tests, por ejemplo).

## Usando el módulo desde otro programa

Una vez que tenemos `matematicas.py` en la carpeta, podemos importarlo desde cualquier otro archivo. Veamos las distintas formas:

````python
# Forma 1: importar el módulo completo
import matematicas

print(matematicas.mcd(48, 18))
print(matematicas.es_primo(7))
````

````python
# Forma 2: importar con alias (muy práctico)
import matematicas as m

print(m.mcd(48, 18))
print(m.es_primo(7))
````

````python
# Forma 3: importar solo lo que necesitas
from matematicas import mcd, es_primo

print(mcd(48, 18))
print(es_primo(7))
````

> ⚠️ **Una forma que NO debes usar**: `from matematicas import *`. Aunque parece cómoda, importa TODOS los nombres del módulo en tu espacio, y puede sobrescribir cosas tuyas silenciosamente. **Importa explícitamente lo que necesites**.

## Reutilizando el módulo: tu primera biblioteca

Mira este programa que usa nuestro módulo para encontrar **el primer primo mayor que un número dado**. Compáralo con lo que habrías escrito en el Tema 4 sin tener `es_primo` como función:

````python
import matematicas as m

n = 100
candidato = n + 1
while not m.es_primo(candidato):
    candidato = candidato + 1

print(f'El primer primo mayor que {n} es {candidato}')
````

**Cuatro líneas**. Eso es todo. Sin la función `es_primo` habrías necesitado el bucle interno de comprobación de divisores, anidado con el bucle exterior. **El módulo te ha permitido pensar a un nivel de abstracción mayor**.
