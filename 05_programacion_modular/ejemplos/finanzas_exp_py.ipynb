# Finanzas: El Valor del Dinero en el Tiempo

¿Cuánto vale hoy recibir 100€ cada año para siempre? Este cuaderno utiliza **series geométricas** para calcular el Valor Presente de una renta perpetua.

## 1. El Concepto de Descuento
El dinero futuro vale menos que el dinero presente debido a la inflación y al interés que podrías ganar si tuvieras el dinero hoy. 

La fórmula estándar en finanzas que permite convertir un valor futuro en un valor presente es la del factor de descuento (*FD*). Este se calcula dividiendo 1 entre el factor de interés compuesto (1+*i*) elevado al número de periodos (*t*), siendo **i** la tasa de interés: 

$$FD=(1+i)^{-t}$$ o $$FD=\frac{1}{(1+i)^t}$$

El factor de descuento el primer año es:
$$r = \frac{1}{1 + i}$$

La suma de todos los pagos futuros es una serie:
$$VP = a + ar + ar^2 + ar^3 + \dots$$

Como $r < 1$, la serie converge a:
$$VP = \frac{a}{1 - r} = \frac{a}{i}$$

```python
# --- CONFIGURACIÓN INICIAL ---
pago_anual = 100  # La cantidad 'a'
tasa_interes = 0.05  # 5% de interés anual

print(f"Pago anual prometido: {pago_anual}€")
print(f"Tasa de interés (descuento): {tasa_interes * 100}%")
```

## 2. Cálculo Interactivo
Ejecuta la siguiente celda para calcular el valor de la renta perpetua. 
**Nota en VS Code:** Mira en la parte superior central de la pantalla para introducir el dato.

```python
try:
    nueva_tasa = float(input("Introduce la tasa de interés deseada (ej. 0.05 para 5%): "))
    
    valor_presente = pago_anual / nueva_tasa
    
    print(f"--- RESULTADO ---")
    print(f"Para una tasa del {nueva_tasa*100:.1f}%, el valor actual de la renta es: {valor_presente:.2f}€")
except ValueError:
    print("Por favor, introduce un número decimal válido (usa el punto para los decimales).")
```

## 3. ¿Por qué converge?
A medida que pasan los años, el valor de ese pago de 100€ se vuelve casi cero. Vamos a ver cuánto vale el pago del año 50 y del año 100 hoy:

```python
def valor_pago_en_año(n, a, i):
    return a / (1 + i)**n

print(f"Valor hoy del pago del año 1: {valor_pago_en_año(1, pago_anual, tasa_interes):.2f}€")
print(f"Valor hoy del pago del año 50: {valor_pago_en_año(50, pago_anual, tasa_interes):.2f}€")
print(f"Valor hoy del pago del año 100: {valor_pago_en_año(100, pago_anual, tasa_interes):.2f}€")
```

### Conclusión
Como puedes ver, el pago del año 100 apenas vale **0.76€** hoy. Por eso, aunque los pagos sean para siempre, la suma total no es infinita. 

Este es el principio fundamental detrás de los **fondos de pensiones** y la valoración de **acciones en bolsa**.
