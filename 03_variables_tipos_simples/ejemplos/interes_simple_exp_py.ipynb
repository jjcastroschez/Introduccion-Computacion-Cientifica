# Simulación Cálculo del interés simple: La mejor opción

En este cuaderno se explica la implentación en Python de la solución al problema de cómo decidir la mejor oferta en base al cálculo del **interés simple**.

## 1. El Problema
Una persona quiere tener un simulador de cálculo del interés simple ofertado para conocer cuál es la mejor opción.

El interés simple se calcula solo sobre el capital inicial durante todo el tiempo, con la fórmula:

$$I = P \times r \times t$$

donde:

I = Interés simple
P = Capital inicial
r = Tasa de interés anual (en decimal)
t = Tiempo (en años)

Necesitamos una variable para almacenar el capital inicial. Para construir su identificador vamos a seguir el estilo **snake_case**, y para que tenga *semántica* decidimos que la cadena `capital_inicial` representa perfectamente el valor que almacena, mejor que P si el día de mañana tenemos que recordar qué es lo que hemos hecho. Esta será una variable de entrada y es de tipo real. 

También hay que almacenar el porcentaje de interés anual que ofrece la entidad bancaria, la vamos a llamar `porcentaje_tasa_interes`, que es el valor que habitualmente publicitan. Esta variable es una variable de entrada de tipo real.

Y por último hay que solicitar el tiempo que hay que tener el dinero en la entidad para poder recuperarlo. Lo almacenaremos en una variable llamada `tiempo`, que será empleada para almacenar el número de años que debe estar el dinero en la entidad sin recuperarse y que será de tipo entero. 

Como variable intermedia vamos a usar la variable `tasa_interes` que la vamos a obtener a partir de otro dato que nos dan en la entrada, que es `porcentaje_tasa_interes` y será de tipo real. 

Con el uso de estos identificadores logramos darle semántica a los nombres y luego, con el tiempo, será más fácil recordar qué era lo que hacen nuestros script o programas.

```python
capital_inicial = float(input("Capital inicial: "))
porcentaje_interes_anual = float(input("Tasa de interés anual (%): "))
tiempo = int(input("Tiempo en años: "))

print(f"Veamos el interés simple para un capital inicial de {capital_inicial}€ con una tasa de interés anual del {porcentaje_interes_anual}% durante {tiempo} años.")
```

## 2. Cálculo del interés simple
Ya hemos presentado la fórmula:
$$I = P \times r \times t$$

Necesitamos una variable, de salida y tipo real, para almacenar el valor del interés simple, la vamos a llamar `interes_simple`, es un valor que estamos calculando para dárselo al usuario. Además, vamos a calcular el monto final, que vamos a almacenar en una variable, también de salida y tipo real, cuyo identificador será `monto_final`:

$$MF= P + I$$

Calculémoslo con Python:

```python
interes_anual = porcentaje_interes_anual / 100
interes_simple = capital_inicial * interes_anual * tiempo
monto_final= capital_inicial+ interes_simple

# Resultados
print(f" \nInterés simple: {interes_simple:.2f}")
print(f"Monto total: {monto_final:.2f}")
```
