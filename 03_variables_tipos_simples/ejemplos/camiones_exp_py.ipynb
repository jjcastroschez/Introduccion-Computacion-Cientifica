# Logística y gestión de inventario: reparto de carga en camiones 

En este cuaderno se explica la implentación en Python de la solución al problema de cómo repartir una carga en camiones con dos capacidades distintas, en él se plantea una ecuación diofántica cuya solución se valida con la **identidad de Bézout**.

## 1. El Problema

Una empresa de logística en España necesita enviar exactamente **31** toneladas de mercancía. Para ello, dispone únicamente de dos tipos de camiones:

 * Camiones grandes con capacidad de *7* toneladas.
 * Camiones medianos con capacidad de *5* toneladas.
 
El gestor de la flota necesita saber cuántos camiones de cada tipo debe enviar para que vayan totalmente llenos, sin desperdiciar espacio ni dejar mercancía en tierra.

Este problema se puede plantear como una **ecuación diofántica** que en caso de verificar la **Identidad de Bézout** se garantizan soluciones en los números enteros (Z), lo que incluye números negativos. Sin embargo, en la vida real (como si estuvieras comprando manzanas y peras), solemos buscar soluciones en los naturales (N), es decir, enteros positivos. Veamos cómo se plantea la ecuación.

## 2. Plantear la ecuación diofántica y verificar solución

Definimos *x* como el número de camiones de **7** toneladas e *y* como el número de camiones de **5** toneladas. Queremos resolver la ecuación:

$$7x+5y=31$$

La **identidad de Bézout** indica que esta ecuación tiene solución en números enteros si y solo si el máximo común divisor de 7 y 5 divide a 31. (*mcd*(7,5)=1). Como 1 divide a 31, la identidad de Bézout garantiza que existen soluciones enteras. Veamos si son positivas.

## 3. Pasos para obtener la solución a la ecuación

Una vez verificado que la ecuación tiene solución, seguimos los siguientes pasos para obtenerla:

1. Trabaja siempre con el número más pequeño de los dos coeficientes para que los cálculos sean fáciles. En este caso 5. Este será el módulo.
2. Simplifica la ecuación dividiéndola por tu módulo (en este caso, entre 5). Desaparece el término del módulo.
3. Calcula el Inverso Modular del valor que acompaña al término que queda. Busca el número que multiplicado por el coeficiente dé resto 1 al dividirlo por el módulo (el 5).
4. Multiplica el resultado simplificado (paso 2) por el inverso (paso 3). Ese será el valor del coeficiente.

En nuestro caso, el módulo sería 5 (Paso 1). La ecuación quedaría como (Paso 2): 

$$2x=1$$

El inverso modular sería 3 (Paso 3), ya que $$3\times2=6$$ y $$6 mod 5 = 1$$ 

El resultado simplificado es 1 (Paso 2) que multiplicado por el inverso calculado (Paso 3) módulo 5 produce el número 3. Luego: $$x=3$$ 

Una vez que tenemos ese valor, simplemente hay que despejar:

$$y=\frac{31-7\times3}{5}=2$$

Nuestra respuesta al problema sería que se necesitan 3 vehículos de capacidad 7 y 2 de capacidad 5 para trasladar la carga 31.

Vamos a solucionarlo de manera general para cualquier otro problema, siempre que se verifique la identidad de Bézout y se permitan soluciones positivas. 

```python
capacidad_mg=int(input("Dame la capacidad del vehículo más grande: "))
capacidad_mp=int(input("Dame la capacidad del vehículo más pequeño: "))
carga=int(input("Dame la carga que quieres distribuir: "))
inverso_cp_mp=int(input(f"Dame el inverso módulo {capacidad_mp} de {capacidad_mg}: "))
```

En una ecuación planteada como:

$$ax+by=r$$

La variable `capacidad_mg` es el coeficiente *a*, `capacidad_mp`sería el coeficiente *b* y *r* sería la `carga`. Estos datos están en el enunciado del problema y serían datos de entrada. Hay un dato `inverso_cp_mp` que es también un dato de entrada pero que debe calcular el gestor de flota, es también una variable de entrada puesto que no se calcula en el script que estamos escribiendo (es el inverso de *a* módulo *b*).  

Calculamos el número de vehículos grandes (x) usando la fórmula del inverso:

```python
num_vehiculos_g = (carga * inverso_cp_mp) % capacidad_mp
print(f"Se necesitan: {num_vehiculos_g} vehículos de capacidad {capacidad_mg} ")
```

Calculamos el número de vehículos pequeños (y) despejando la ecuación original: y = (c - ax) / b

```python
num_vehiculos_p = (carga - capacidad_mg * num_vehiculos_g) // capacidad_mp
print(f"Se necesitan: {num_vehiculos_p} vehículos de capacidad {capacidad_mp} ")
```

Proporcionamos la salida al gestor de la flota:

```python
print(f"Se necesitan {num_vehiculos_g} vehículos de capacidad de {capacidad_mg} y {num_vehiculos_p} vehículos de capacidad {capacidad_mp} para transportar una carga de {carga}.")
```
