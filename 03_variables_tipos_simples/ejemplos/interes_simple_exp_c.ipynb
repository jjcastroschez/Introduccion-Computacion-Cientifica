# Simulación Cálculo del interés simple: La mejor opción

En este cuaderno se explica la implentación en C de la solución al problema de cómo decidir la mejor oferta en base al cálculo del **interés simple**.

## 1. El Problema
Una persona quiere tener un simulador de cálculo del interés simple ofertado para conocer cuál es la mejor opción.

El interés simple se calcula solo sobre el capital inicial durante todo el tiempo, con la fórmula:

$$I = P \times r \times t$$

donde:

I = Interés simple
P = Capital inicial
r = Tasa de interés anual (en decimal)
t = Tiempo (en años)

### Configuración de la Simulación
En C hay que declarar las variables indicando su tipos correspondientes (`int`, `float`, `char`) antes de usarlas (**declaración explicita**). Usaremos dos funciones que están definidas en el lenguaje, en la librería `<stdio.h>`, de esto hablaremos en el Tema 5 (por eso hacemos: `#include <stdio.h>`). Las funciones son `printf` para mostrar mensajes y resultados por pantalla, y `scanf` que nos permite leer datos de la entrada estándar (`stdin`, usualmente el teclado) y almacenarlos en variables.

La sintaxis básica de uso de `scanf`es:
    `scanf("especificador_formato", &variable)`

El operador `&` es obligatorio para los tipos básicos (int, float, char) para indicar a `scanf` dónde guardar el dato. 

Los especificadores de formato válidos son:


| Especificador | Tipo de Dato                    | 
|---------------|---------------------------------|
| %d            | Entero decimal                  | 
| %f            | Número en coma flotante (float) | 
| %c            | Carácter, uno solo (char)       |
| %s            | Cadena de caracteres (string)   |

Cuidado con el especificador %s, puede ser inseguro si no se limita la longitud. Además, `scanf` leerá caracteres hasta encontrar un espacio, tabulación o salto de línea.

Las variables que vamos a usar son las mismas que ya dijimos en el ejercicio [interes_simple_exp_py.ipynb](/03_variables_tipos_simples/ejemplos/interes_simple_exp_py.ipynb). 

Recuerda que vamos a usar una variable, de salida y tipo real, para almacenar el valor del interés simple, la vamos a llamar `interes_simple`, es un valor que estamos calculando para dárselo al usuario. Además, vamos a calcular el monto final, que vamos a almacenar en una variable, también de salida y tipo real, cuyo identificador será `monto_final`. Estas variables también tienen que ser declaradas.

````cpp
#include <stdio.h>    

    // --- DECLARACION DE VARIABLES ---
    
    float capital_inicial;
    float porcentaje_interes_anual;
    int  tiempo;

    float interes_anual;
    float interes_simple;
    float monto_final;

    // --- ENTRADA DE DATOS -- 

    printf("Capital inicial: ");
    scanf("%f", &capital_inicial);
    printf("Tasa de interés anual (%%): ");
    scanf("%f", &porcentaje_interes_anual);
    printf("Tiempo en años: ");
    scanf("%d", &tiempo);   
````

## 2. Cálculo del interés simple
Ya hemos presentado la fórmula del interés simple:
$$I = P \times r \times t$$

Y calcularemos el monto final conforme a la siguiente fórmula:

$$MF= P + I$$

Ahora hacemos los cálculos con C:

````cpp
// --- CÁLCULOS INICIALES ---
    
    interes_anual = porcentaje_interes_anual / 100.0;
    interes_simple = capital_inicial * interes_anual * tiempo;  
    monto_final = capital_inicial + interes_simple;
    

    // --- SALIDA DE RESULTADOS ---
    printf("\nInterés simple: %.2f\n", interes_simple);
    printf("Monto total: %.2f\n", monto_final);
````
