# Finanzas: El Valor del Dinero en el Tiempo

¿Cuánto vale hoy recibir 100€ cada año para siempre? Este cuaderno utiliza **series geométricas** para calcular el Valor Presente de una renta perpetua.

## 1. El Concepto de Descuento
El dinero futuro vale menos que el dinero presente debido a la inflación y al interés que podrías ganar si tuvieras el dinero hoy. 

La fórmula estándar en finanzas que permite convertir un valor futuro (VF) en un valor presente (VP) es la del factor de descuento (*FD*). Este se calcula dividiendo 1 entre el factor de interés compuesto (1+*i*) elevado al número de periodos (*t*), siendo **i** la tasa de interés: 

$$FD=(1+i)^{-t}$$ o $$FD=\frac{1}{(1+i)^t}$$

El factor de descuento el primer año es:
$$r = \frac{1}{1 + i}$$

La suma de todos los pagos futuros es una serie:
$$VP = a + ar + ar^2 + ar^3 + \dots$$

Como $r < 1$, la serie converge a:
$$VP = \frac{a}{1 - r} = \frac{a}{i}$$

```python
pago_anual = 100  # La cantidad 'a'
tasa_interes = 0.05  # 5% de interés anual

print(f"Pago anual prometido: {pago_anual}€")
print(f"Tasa de interés (descuento): {tasa_interes * 100}%")
```

## 2. Cálculo Valor de Presente
Ejecuta la siguiente celda para calcular el valor presente de la renta ofrecida. 

```python
valor_perpetuidad = pago_anual / tasa_interes

print(f"Para una tasa del {tasa_interes*100:.1f}%, recibiendo {pago_anual}€ anuales, el valor presente ofrecido es: {valor_perpetuidad:.2f}€")
```

## 3. ¿Por qué converge?
A medida que pasan los años, el valor de ese pago de 100€ se vuelve casi cero. Vamos a ver cuánto vale el pago de los años 5, 10, 25, 50 y 100 hoy:

```python
print(f"Valor hoy del pago del año 5: {(pago_anual/ (1+tasa_interes)**5):.2f}€")
print(f"Valor hoy del pago del año 10: {(pago_anual/ (1+tasa_interes)**10):.2f}€")
print(f"Valor hoy del pago del año 25: {(pago_anual/ (1+tasa_interes)**25):.2f}€")
print(f"Valor hoy del pago del año 50: {(pago_anual/ (1+tasa_interes)**50):.2f}€")
print(f"Valor hoy del pago del año 100: {(pago_anual/ (1+tasa_interes)**100):.2f}€")
```

Como puedes observar el dinero cada vez vale menos... Fíjate que en el año 100 apenas vale 0.76€ de hoy. Por eso, aunque los pagos sean para siempre, la suma total no es infinita.

## 4. ¿Cuánto tengo un año en concreto?
Si ahorras cada pago de 100€ (*a*) y estos generan un interés del 0.05% (*i*) cada año, la fórmula para el acumulado en el año *k* será:

$$VF =a \cdot \frac{(1+i)^k -1}{i}$$

En la siguiente celda del cuaderno calculamos el acumulado en cualquier año, se lo solicitamos al usuario.

```python
k = int(input("Introduce el año k para ver el acumulado: "))
valor_futuro = (pago_anual * ((1 + tasa_interes)**k - 1)) / tasa_interes

print(f"--- CAPITAL ACUMULADO ---")
print(f"En el año {k}, tras reinvertir los pagos al {tasa_interes*100}%, tendrás: {valor_futuro:.2f}€")
```

### Conclusión

Este es el principio fundamental detrás de los **fondos de pensiones** y la valoración de **acciones en bolsa**.
