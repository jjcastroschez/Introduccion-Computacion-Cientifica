# Simulación Farmacológica: El Estado Estacionario (Versión C)

En este cuaderno se explica la implementación en C de la solución al problema de cómo la medicina se acumula en el cuerpo hasta alcanzar un equilibrio, utilizando el concepto de **series geométricas convergentes**.

## 1. El Problema
Un paciente toma una dosis diaria **$a$**. El cuerpo elimina un porcentaje cada día, por lo que sobrevive una fracción **$r$**.

La cantidad de medicamento en el cuerpo justo después de cada dosis es:
* **Día 1:** $a$
* **Día 2:** $a + a(r)$
* **Día 3:** $a + a(r) + a(r^2)$

Esto forma una serie geométrica: $S = a \sum_{n=0}^{k-1} r^n$

### Configuración de la Simulación
En C, recuerda que hay que declarar las variables indicando su tipos correspondientes (`int`, `float`, `char`) antes de usarlas (**declaración explicita**). Usamos `printf` para mostrar mensajes y resultados por pantalla. Y la forma de leer datos de la entrada estándar (`stdin`, usualmente el teclado) y almacenarlos en variables se hará empleando la función `scanf`. Estas funciones requieren del uso de la librería `<stdio.h>`, de esto hablaremos en el Tema 5 (por eso hacemos: `#include <stdio.h>`).

Recuerda que la sintaxis básica de uso de `scanf` es:
    `scanf("especificador_formato", &variable)`

Mira el ejercicio [interes_simpre_exp_c.ipynb](/03_variables_tipos_simples/ejemplos/interes_simple_exp_c.ipynb) para más información sobre el uso de estas funciones.

Cuando queramos hacer este programa general, habrá que solicitar al médico la `dosis_diaria` y el `porcentaje_eliminado`, esto se hará a través de las sentencias:

```c
printf("\nIntroduce la cantidad diaria de medicamento: ");
scanf("%d", &dosis_diaria); 
print("Introduce el porcentaje eliminado diariamente del medicamento (0 a 1): ");
scanf("%f", &porcentaje_eliminado);
``` 

Con los datos del caso concreto: toma de 100 mg de medicamento diario y 0,8 de porcentaje eliminado diariamente:

````cpp
#include <stdio.h>

// --- CONFIGURACIÓN DE LA SIMULACIÓN ---
int dosis_diaria = 100;           // mg
float porcentaje_eliminado = 0.2;
float r = 1.0f - porcentaje_eliminado; // Fracción que permanece (0.8)

printf("Dosis: %d mg\n", dosis_diaria);
printf("Factor de permanencia (r): %.2f\n", r);
````

Observa la declaración de una junto con su inicialización, que puede o no implicar un cálculo.

## 2. Cálculo de la Convergencia
A largo plazo ($k \to \infty$), la cantidad de medicamento converge a:
$$S_\infty = \frac{a}{1 - r}$$

````cpp
float limite_estacionario = dosis_diaria / (1.0f - r);

printf("A largo plazo, el paciente tendrá un nivel estable de: %.2f mg\n", limite_estacionario);
printf("Este es el 'Estado Estacionario'.\n");
````

Observa que la declaración de una variable se puede hacer en culquir sitio, pero siempre antes de usarla. Es una buena práctica declarar todas las variables que se vayan a usar al comienzo. 

## 3. Cantidad de medicamento en la sangre
En cualquier momento podremos conocer cuál será la acumulación del fármaco en el tratamiento:
$$S_k = a \left(\frac{1-r^k}{1 - r}\right)$$

Para elevar a una potencia en C, necesitamos la librería `<math.h>` y la función `pow()`.

````cpp
#include <math.h>

int dia_tratamiento = 15;
float cantidad_en_cuerpo = dosis_diaria * (1.0f - pow(r, dia_tratamiento)) / (1.0f - r);

printf("La cantidad de medicamento el día %d es %.2f mg\n", dia_tratamiento, cantidad_en_cuerpo);
````

## 4. Experimenta

En la generalización, para permitir que el usuario interactúe, no puedes usar `scanf`, que es lo que usaremos en el programa que hemos desarrollado. Nota: `scanf`está pensado para programas que se ejecutan en una **terminal interactiva**. Pero un notebook de Jupyter **no tiene terminal interactiva conectada al kernel de C**: la celda se ejecuta, el kernel intenta leer de stdin, y como no hay nadie escribiendo... o se queda colgado esperando indefinidamente, o lee basura/EOF y `dia_usuario` queda con un valor indeterminado.

En el programa en C, lo haremos así:

```c
int dia_usuario;
printf("Introduce el día de tratamiento: ");
scanf("%d", &dia_usuario);

float cantidad_usuario = dosis_diaria * (1.0f - pow(r, dia_usuario)) / (1.0f - r);
printf("La cantidad de medicamento el día %d es %.2f mg\n", dia_usuario, cantidad_usuario);
```

Pero en los notebooks con código C/C++ la solución más sencilla es no usar `scanf` y asignar el valor con el que se quiere experimentar directamente a la variable:

```c
dia_tratamiento=22
```

````cpp
dia_tratamiento=22;

cantidad_en_cuerpo = dosis_diaria * (1.0f - pow(r, dia_tratamiento)) / (1.0f - r);
printf("La cantidad de medicamento el día %d es %.2f mg\n", dia_tratamiento, cantidad_en_cuerpo);
````
