# Simulación Farmacológica: El Estado Estacionario

En este cuaderno se explica la implentación en Python de la solución al problema de cómo la medicina se acumula en el cuerpo hasta alcanzar un equilibrio, utilizando el concepto de **series geométricas convergentes**.

## 1. El Problema
Un paciente toma una dosis diaria **$a$**. El cuerpo elimina un porcentaje cada día, por lo que sobrevive una fracción **$r$**.

La cantidad de medicamento en el cuerpo justo después de cada dosis es:
* **Día 1:** $a$
* **Día 2:** $a + a(r)$
* **Día 3:** $a + a(r) + a(r^2)$

Esto forma una serie geométrica: $S = a \sum_{n=0}^{k-1} r^n$

Necesitamos una variable para almacenar la dosis diaria. Para construir su identificador vamos a seguir el estilo **snake_case**, y para que tenga *semántica* decidimos que la cadena `dosis_diaria` representa perfectamente el valor que almacena. Esta será una variable de entrada y es de tipo entero. 

También necesitamos la razón de crecimiento de la serie, la vamos a llamar `r`, que está ampliamente usado y aceptado en el campo de las matemáticas cuando trabajamos con series geométricas. 

Esta variable `r`, es una variable intermedia de tipo real y la vamos a obtener a partir de otro dato que nos dan en la entrada, que es el porcentaje eliminado del medicamento diariamente, este dato lo vamos a almacenar en una variable cuyo identificador será `porcentaje_eliminado` y será de tipo real. 

Ambas variables de entrada las vamos a incluir en el script, no es necesario pedirlas a usuario para este caso concreto, si quisieramos hacer un script general deberíamos pedirlas al usuario al comenzar la ejecución del script, con las sentencias 

```python
dosis_diaria = int(input("Introduce la cantidad diaria de medicamento: "))  # mg
porcentaje_eliminado = float(input("Introduce el porcentaje eliminado diariamente del medicamento (0 a 1): "))
``` 

````python
# --- CONFIGURACIÓN DE LA SIMULACIÓN ---
dosis_diaria = 100  # mg
porcentaje_eliminado = 0.20
r = 1 - porcentaje_eliminado  # Fracción que permanece (0.8)

print(f"Dosis: {dosis_diaria} mg")
print(f"Factor de permanencia (r): {r}")
````

## 2. Cálculo de la Convergencia
A largo plazo ($k \to \infty$), la cantidad de medicamento no sube para siempre, sino que converge a:
$$S_\infty = \frac{a}{1 - r}$$

Calculémoslo con Python:

Necesitamos una variable para almacenar el valor de convergencia, en una variable de salida, puesto que es un valor que estamos calculando para dárselo al usuario. A esta variable la vamos a llamar `limite_estacionario` y será de tipo real.  

````python
limite_estacionario = dosis_diaria / (1 - r)

print(f"A largo plazo, el paciente tendrá un nivel estable de: {limite_estacionario} mg")
print(f"Este es el 'Estado Estacionario'.")
````

## 3. Cantidad de medicamento en la sangre
En cualquier momento podremos conocer cuál será la acumulación del fármaco en el tratamiento:
$$S_k = a \left(\frac{1-r^k}{1 - r}\right)$$

Calculemos en Python cuál es la acumulación de medicamento en la sangre, por ejemplo al décimo quinto día de comenzar el tratamiento:

Este podría considerarse otro script, puesto que contesta a otra pregunta, pero podemos también incluirlo dentro del mismo como otra funcionalidad. Puesto que el día por el que nos preguntan es el día 15, este debe almacenarse en alguna variable para poderse usar en los cálculos del script. Vamos a usar una varible de tipo entero cuyo identificador será `dia_tratamiento`.  Deberá estar inicializado en código al valor que nos interesa 15. 

Además, necesitamos calcular la variable de salida que nos informa de la cantidad de medicamento que hay en el cuerpo el dia 15, esa variable será de tipo real, y para su identificador hemos decidido que sea `cantidad_en_cuerpo`.

````python
dia_tratamiento=15
cantidad_en_cuerpo = dosis_diaria * (1-r**dia_tratamiento)/(1-r)
print(f"La cantidad de medicamento el día {dia_tratamiento} es {cantidad_en_cuerpo:.2f} mg")
````
