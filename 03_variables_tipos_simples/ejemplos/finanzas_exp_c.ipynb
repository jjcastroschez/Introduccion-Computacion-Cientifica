## 1. El Concepto de Descuento
El dinero futuro vale menos que el dinero presente debido a la inflación y al interés que podrías ganar si tuvieras el dinero hoy. 

La fórmula estándar en finanzas que permite convertir un valor futuro (VF) en un valor presente (VP) es la del factor de descuento (*FD*). Este se calcula dividiendo 1 entre el factor de interés compuesto (1+*i*) elevado al número de periodos (*t*), siendo **i** la tasa de interés: 

$$FD=(1+i)^{-t}$$ o $$FD=\frac{1}{(1+i)^t}$$

El factor de descuento el primer año es:
$$r = \frac{1}{1 + i}$$

La suma de todos los pagos futuros es una serie:
$$VP = a + ar + ar^2 + ar^3 + \dots$$

Como $r < 1$, la serie converge a:
$$VP = \frac{a}{1 - r} = \frac{a}{i}$$

Recuerda que en C la declaración de las variables es explícita, debes declarar una variable antes de usarla indicando el tipo que tiene. Este tipo no puede cambiar a lo largo del programa. 

```c++
#include <stdio.h>

// --- DECLARACION DE VARIABLES ---
float pago_anual;
float tasa_interes;
float valor_perpetuidad;
float valor_futuro;
int k;  

// --- ENTRADA DE DATOS --- 

pago_anual = 100;  // La cantidad 'a'
tasa_interes = 0.05;  // 5% de interés anual

printf("\nPago anual prometido: %.2f€", pago_anual);
printf("\nTasa de interés (descuento): %.1f%%", tasa_interes * 100);
```

## 2. Cálculo Valor de Presente
Ejecuta la siguiente celda para calcular el valor presente de la renta ofrecida. 

```c++
    valor_perpetuidad = pago_anual / tasa_interes;

    printf("\nPara una tasa del %.1f%%, recibiendo %.2f€ anuales, el valor a perpetuidad es: %.2f€", tasa_interes * 100, pago_anual, valor_perpetuidad);
```

## 3. ¿Por qué converge?
Vamos a ver como comprobamos en C que a medida que pasan los años, el valor de ese pago de 100€ se vuelve casi cero. Es interesante como tenemos que hacer uso de `#include <math>`para poder usar la función `pow` esto ya lo enterás mejor en el Tema 5. Si quieres prueba a eliminarlo en tu copia y mira lo que ocurre. Veamos cuánto vale el pago de los años 5, 10, 25, 50 y 100 hoy:

```c++
#include <math.h>

printf("\nValor hoy del pago del año 5: %.2f€\n", (pago_anual/ pow((1+tasa_interes), 5)));
printf("Valor hoy del pago del año 10: %.2f€\n", (pago_anual/ pow((1+tasa_interes), 10)));
printf("Valor hoy del pago del año 25: %.2f€\n", (pago_anual/ pow((1+tasa_interes), 25)));
printf("Valor hoy del pago del año 50: %.2f€\n", (pago_anual/ pow((1+tasa_interes), 50)));
printf("Valor hoy del pago del año 100: %.2f€\n", (pago_anual/ pow((1+tasa_interes), 100)));
```

Como puedes observar el dinero cada vez vale menos... Fíjate que en el año 100 apenas vale 0.76€ de hoy. Por eso, aunque los pagos sean para siempre, la suma total no es infinita.

## 4. ¿Cuánto tengo un año en concreto?
Si ahorras cada pago de 100€ (*a*) y estos generan un interés del 0.05% (*i*) cada año, la fórmula para el acumulado en el año *k* será:

$$VF =a \cdot \frac{(1+i)^k -1}{i}$$

En la siguiente celda del cuaderno calculamos el acumulado en cualquier año, este dato se lo solicitamos al usuario. Observa que el año lo almacenamos en la variable `k`, sin duda alguna más semántica tendría una variable que se llamara `año` pero recuerda que no es aconsejable usar variables que tengan caracteres extraños. La *ñ* no es un elemento del abecedario inglés, por lo que muchos lenguajes no la permitirán en la construcción de identificadores. Podríamos usar el latín y poner al identificador `annus` o dejar el `k` e indicar en los comentarios que se trata del *año*. 

```c++
printf("\nIntroduce el año k para ver el acumulado: ");
scanf("%d", &k);

valor_futuro = (pago_anual * (pow((1 + tasa_interes),k - 1))) / tasa_interes;

printf("--- CAPITAL ACUMULADO ---\n");
printf("En el año %d, con un pago de %.2f€ al %.1f%%, tendrás:\n", k, pago_anual, tasa_interes * 100);
printf("%.2f€\n", valor_futuro);
```
