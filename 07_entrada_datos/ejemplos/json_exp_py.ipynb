**JSON** (*JavaScript Object Notation*) es el formato dominante para intercambiar datos entre programas modernos, sobre todo con **APIs web**. Su sintaxis es prácticamente idéntica a los diccionarios y listas de Python:

```json
{
  "titulo": "Bowie",
  "anio": 2019,
  "temas": ["biografía", "música"]
}
```

El módulo `json` de Python convierte entre JSON y estructuras nativas de Python en una sola línea.

## 1. Leer un JSON con `json.load()`

`json.load(fichero)` lee el JSON del fichero y lo convierte en la estructura Python equivalente:

| JSON | Python |
|:---|:---|
| objeto `{...}` | dict |
| array `[...]` | list |
| string `"..."` | str |
| number `42` o `3.14` | int o float |
| `true` / `false` | True / False |
| `null` | None |

Es un mapeo casi perfecto.

````python
import json

with open('libro.json', 'r') as f:
    datos = json.load(f)

print(f"Tipo: {type(datos).__name__}")
print(f"Título:   {datos['titulo']}")
print(f"Autor:    {datos['autor']}")
print(f"Año:      {datos['anio']}")
print(f"Páginas:  {datos['paginas']}")
````

Fíjate en que los **tipos se preservan**: los números vienen como números, no como strings; las listas vienen como listas de Python, no como cadenas. Es una gran ventaja frente a CSV.

````python
# 'temas' es una LISTA de Python (no una cadena)
print(f"Tipo de 'temas': {type(datos['temas']).__name__}")
print(f"Cantidad: {len(datos['temas'])}")

for tema in datos['temas']:
    print(f'  • {tema}')

# 'disponible' es un BOOLEANO
print(f"\nTipo de 'disponible': {type(datos['disponible']).__name__}")
print(f'¿Disponible? {datos["disponible"]}')
````

## 2. Escribir un JSON con `json.dump()`

`json.dump(datos, fichero)` hace la conversión inversa: toma un dict/lista de Python y lo escribe en el fichero como JSON.

Parámetros útiles:

* **`indent=2`** — formatea el JSON con indentación de 2 espacios. Sin él, todo va en una única línea larga.
* **`ensure_ascii=False`** — permite guardar caracteres UTF-8 tal cual (tildes, ñ, €...). Sin él, `Cálculo` se guarda como `C\u00e1lculo`.

````python
libro = {
    'titulo': 'Cálculo Infinitesimal',
    'autor': 'Michael Spivak',
    'anio': 1967,
    'temas': ['cálculo', 'análisis', 'matemáticas'],
}

with open('mi_libro.json', 'w', encoding='utf-8') as f:
    json.dump(libro, f, indent=2, ensure_ascii=False)

print('✅ mi_libro.json guardado.\n')

# Comprobar el resultado
with open('mi_libro.json') as f:
    print(f.read())
````

## 3. Conversión sin fichero: `dumps` y `loads`

A veces tienes el JSON en una **cadena** (por ejemplo, la respuesta de una API) y no en un fichero. Para esos casos hay dos primos hermanos con **`s`** al final:

* **`json.dumps(datos)`** — convierte dict/lista Python en **string** JSON.
* **`json.loads(string)`** — convierte string JSON en dict/lista Python.

Regla mnemotécnica: la `s` es de **string** (o de "string version").

````python
# Python → string JSON
s = json.dumps(libro, ensure_ascii=False)
print(f'Tipo: {type(s).__name__}')
print(f'Contenido: {s}')
print()

# String JSON → Python
d = json.loads(s)
print(f'De vuelta a: {type(d).__name__}')
print(f"Título: {d['titulo']}")
````

## 4. Datos anidados

JSON brilla cuando los datos son **estructurados y anidados**. Por ejemplo, una lista de libros donde cada libro tiene autores múltiples:

````python
biblioteca = {
    'nombre': 'Biblioteca Facultad de Matemáticas',
    'libros': [
        {'titulo': 'Cálculo', 'autores': ['Michael Spivak'], 'anio': 1967},
        {'titulo': 'Álgebra Lineal', 'autores': ['Kenneth Hoffman', 'Ray Kunze'], 'anio': 1971},
        {'titulo': 'Análisis Real', 'autores': ['Walter Rudin'], 'anio': 1976},
    ],
}

with open('biblioteca.json', 'w', encoding='utf-8') as f:
    json.dump(biblioteca, f, indent=2, ensure_ascii=False)

# Lo leemos y accedemos a lo anidado
with open('biblioteca.json') as f:
    datos = json.load(f)

print(f"Biblioteca: {datos['nombre']}")
for libro in datos['libros']:
    autores = ', '.join(libro['autores'])
    print(f"  📚 {libro['titulo']} ({libro['anio']}) — {autores}")
````

## 5. CSV vs JSON: ¿cuándo usar cada uno?

| Situación | Elección |
|:---|:---|
| Los datos forman una **tabla plana** (siempre las mismas columnas) | **CSV** |
| Quieres abrirlos en Excel | **CSV** |
| Los datos son **jerárquicos** (un libro tiene autores, una empresa tiene empleados...) | **JSON** |
| Los datos tienen **tipos mixtos** (números, listas, booleanos) | **JSON** |
| Vas a intercambiar con una **API web** | **JSON** casi seguro |

Para datos científicos simples (tabla de medidas), CSV es más compacto y directo. Para intercambio moderno, JSON es el estándar.

## 🎯 Resumen: los 4 métodos del módulo json

```python
# Con fichero
datos = json.load(f)              # fichero → Python
json.dump(datos, f, indent=2)     # Python → fichero

# Con cadena
datos = json.loads(cadena)         # string → Python
cadena = json.dumps(datos)         # Python → string
```

Recuerda: **con `s`** = con string. **Sin `s`** = con fichero.
