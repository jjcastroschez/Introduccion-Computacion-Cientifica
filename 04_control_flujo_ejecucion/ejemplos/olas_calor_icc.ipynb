# 🌡️ Detección de olas de calor*(versión explicada paso a paso)*---## El problemaLa **AEMET** (Agencia Estatal de Meteorología) define una **ola de calor** como un episodio en el que la temperatura máxima diaria supera el umbral del **percentil 95** de la serie histórica durante **al menos 3 días consecutivos**.En Castilla-La Mancha, ese umbral ronda los **36 ºC** en pleno verano. Vamos a programar un detector que, dada la lista de temperaturas máximas de un mes, cuente cuántas olas ha habido y cuántos días han durado en total.## Los datosTomamos como ejemplo las temperaturas máximas de Ciudad Real durante un mes ficticio:

```python
temperaturas = [28, 31, 33, 35, 37, 38, 36, 34, 32, 30, 29, 31, 36, 38, 39, 41, 40, 37, 35, 33, 31, 29, 30, 32, 34, 36, 37, 38, 39, 36]print(f'Datos: {len(temperaturas)} días.')
```

Visualmente vemos a ojo dos posibles olas: los días 4-6 (37, 38, 36) y los días 13-19 (36, 38, 39, 41, 40, 37, 35). La segunda es claramente más larga e intensa. Vamos a programar un detector que las cuente automáticamente.

## El patrón: detección de rachasEsto es un caso clásico de un patrón llamado **detección de rachas (*runs detection*)**. La idea es:1. Llevar un **contador de días consecutivos** (`racha`) que se incrementa cuando se cumple la condición y se **reinicia a 0** cuando no.2. Cada vez que la racha alcance el mínimo (3 días), contar **una ola más**.3. Si la racha continúa, sumar también esos días al total.## Paso 1: Definir constantesPara que el código sea **legible y fácil de modificar**, ponemos los valores umbral en mayúsculas (convención Python para constantes).

```python
UMBRAL = 36       # ºC. A partir de esta temperatura consideramos día calurosoDIAS_MIN = 3      # Mínimo de días consecutivos para considerar 'ola de calor'
```

## Paso 2: Inicializar contadores> ⚠️ **Antes de empezar un bucle, inicializa siempre los contadores que vas a usar dentro.** Si te olvidas, Python lanzará un `NameError`. Es uno de los errores más comunes en principiantes.

```python
olas = 0          # Cuántas olas de calor hemos detectadodias_olas = 0     # Cuántos días en total están dentro de olasracha = 0         # Contador de días consecutivos por encima del umbral
```

## Paso 3: Recorrer las temperaturasUsamos un `for` para iterar sobre la lista de temperaturas. Por cada día decidimos:* Si supera el umbral → incrementamos la racha.* Si no → reiniciamos la racha a 0.Además, justo cuando la racha llega exactamente a 3, **acabamos de detectar una nueva ola** (con sus 3 primeros días). Si la racha sigue creciendo (4, 5, 6...), cada día adicional se cuenta para `dias_olas`.

```python
for t in temperaturas:    if t > UMBRAL:        racha = racha + 1        if racha == DIAS_MIN:            # Acabamos de cumplir el mínimo: nueva ola            olas = olas + 1            dias_olas = dias_olas + DIAS_MIN        elif racha > DIAS_MIN:            # Día adicional dentro de una ola ya contada            dias_olas = dias_olas + 1    else:        racha = 0   # se rompe la rachaprint(f'Olas detectadas: {olas}')print(f'Días totales dentro de olas: {dias_olas}')
```

## Paso 4: Visualizar lo que está pasandoUna técnica muy útil para **entender** un bucle (y para depurarlo) es añadir `print()` dentro que nos muestren el estado en cada vuelta. Vamos a hacerlo:

```python
olas = 0dias_olas = 0racha = 0for dia, t in enumerate(temperaturas, start=1):    if t > UMBRAL:        racha = racha + 1        marca = ' 🔥' if racha >= DIAS_MIN else ''        if racha == DIAS_MIN:            olas = olas + 1            dias_olas = dias_olas + DIAS_MIN            marca = ' 🌋 INICIO OLA'        elif racha > DIAS_MIN:            dias_olas = dias_olas + 1        print(f'Día {dia:2}: {t} ºC  (racha={racha}){marca}')    else:        if racha > 0:            print(f'Día {dia:2}: {t} ºC  (corta la racha)')        racha = 0print(f'\nResumen: {olas} olas, {dias_olas} días en olas.')
```

## 🎯 Conceptos del Tema 4 que has practicado* ✅ **Bucle `for`** recorriendo una secuencia (lista de temperaturas).* ✅ **Condicionales anidados** (`if` dentro de `for` y otro `if` dentro del primero).* ✅ **Variables contadoras** correctamente inicializadas antes del bucle.* ✅ Patrón clásico de **detección de rachas** (*runs detection*).* ✅ Uso de **`enumerate()`** para llevar el índice junto al valor (útil tanto en código de producción como para depurar).## 🚀 Para ir más allá* ¿Cómo lo modificarías para detectar también **olas de frío** (3 días consecutivos por debajo de 0 ºC)?* ¿Cómo lo cambiarías para que devuelva además **la duración de la ola más larga**?* ¿Y para que muestre las **fechas exactas de inicio y fin** de cada ola?
