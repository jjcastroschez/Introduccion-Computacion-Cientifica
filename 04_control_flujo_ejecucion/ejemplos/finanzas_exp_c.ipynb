Vamos a ver la implementación en **C** del cálculo del valor presente de una anualidad finita. Es un ejemplo casi idéntico al de farmacología, pero con una sutileza: usamos la función matemática `pow` para elevar a una potencia, y eso obliga a enlazar la **biblioteca matemática** durante la compilación. Esto acabarás de entenderlo en epl Tema 5.

## El código en C

```c
/*
 * Valor presente de una anualidad finita.
 * Tema 4 - Introducción a la Computación Científica (ICC).
 *
 * Compilar con: gcc -o finanzas finanzas.c -lm
 * Ejecutar con: ./finanzas
 *
 * Nota: -lm es necesario para enlazar la biblioteca matemática (pow).
 */

#include <stdio.h>
#include <math.h>

int main(void) {
    double pago_anual, tasa_interes, i, valor_presente;
    int annus, t;

    printf("Pago anual (€): ");
    scanf("%lf", &pago_anual);
    printf("Tasa de interés anual (en %%): ");
    scanf("%lf", &tasa_interes);
    printf("Años de la anualidad: ");
    scanf("%d", &annus);

    i = tasa_interes / 100.0;
    valor_presente = 0.0;

    for (t = 1; t <= annus; t++) {
        valor_presente += pow(1.0 + i, t);   /* error a propósito: ¿lo ves? */
    }
    /* La línea correcta sería:
       valor_presente += pago_anual / pow(1.0 + i, t);
    */

    printf("\nValor presente de la anualidad de %d años: %.2f €\n",
           annus, valor_presente);
    printf("Valor de la renta perpetua (Tema 3): %.2f €\n", pago_anual / i);

    return 0;
}
```

## Diferencias clave con Python

Aparte de las que ya viste en el ejemplo de farmacología (tipado, `main`, `scanf`/`printf`, llaves...), aquí aparecen dos novedades:

### 1. La biblioteca matemática `<math.h>`

Para usar funciones matemáticas (`pow`, `sqrt`, `sin`, `cos`, `log`...), tienes que incluirla con `#include <math.h>` y compilar con la opción `-lm` ("link math"):

```bash
gcc -o finanzas finanzas_icc.c -lm
```

Si olvidas el `-lm`, el compilador te dará un error del tipo `undefined reference to 'pow'`. ¡Acuérdate!

### 2. Escapar el símbolo `%` en `printf`

Como `%` es el carácter especial que introduce los formatos (`%d`, `%lf`...), si quieres imprimir un `%` literal debes duplicarlo:

```c
printf("Tasa de interés anual (en %%): ");
```

En Python no hay nada de esto: el `%` es un carácter normal.
