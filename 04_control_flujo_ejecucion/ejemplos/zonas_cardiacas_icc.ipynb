# 🫀 Zonas de entrenamiento por frecuencia cardíaca*(versión explicada paso a paso)*---## El problemaCuando hacemos deporte, nuestro corazón late más rápido. Los entrenadores y médicos deportivos clasifican el esfuerzo en **5 zonas** según el porcentaje de la **Frecuencia Cardíaca Máxima (FCM)** al que estamos trabajando:| Zona | % de FCM | Efecto || :--- | :---: | :--- || Z1 | 50-60% | Calentamiento, recuperación || Z2 | 60-70% | Quema de grasa, base aeróbica || Z3 | 70-80% | Mejora del sistema cardiovascular || Z4 | 80-90% | Umbral anaeróbico, mejora rendimiento || Z5 | 90-100% | Esfuerzo máximo, no sostenible |La **FCM** se estima con la fórmula de Tanaka *et al.* (2001), que es más precisa que la antigua $220 - \text{edad}$:$$\text{FCM} = 208 - 0.7 \cdot \text{edad}$$

## Paso 1: Pedir los datos al usuarioNecesitamos dos datos: la edad y las pulsaciones actuales. Como `input()` siempre devuelve una cadena de texto, la convertimos a entero con `int()`.

```python
edad = int(input('Edad: '))ppm = int(input('Pulsaciones actuales (ppm): '))
```

> ⚠️ **Cuidado**: si el usuario teclea letras en lugar de números, `int()` lanzará un `ValueError`. En el [Ejemplo 4](validacion_entrada_icc.ipynb) veremos cómo gestionarlo correctamente con `try-except`. Por ahora, asumimos que el usuario teclea bien.

## Paso 2: Calcular la FCM y el porcentaje de esfuerzoAplicamos la fórmula y calculamos el porcentaje de la FCM al que estamos trabajando.

```python
fcm = 208 - 0.7 * edadporcentaje = (ppm / fcm) * 100print(f'Tu FCM teórica es {fcm:.0f} ppm')print(f'Estás trabajando al {porcentaje:.1f}% de tu FCM')
```

## Paso 3: Clasificar el esfuerzo en una zonaAquí entra en juego el **`if-elif-else`**. Tenemos 5 zonas (más un par de casos extremos), así que usamos varias ramas excluyentes.> 💡 **Truco pedagógico**: como las condiciones se evalúan en orden, cuando llegamos al `elif porcentaje < 60` ya sabemos que `porcentaje >= 50` (porque si no, habríamos entrado en el primer `if`). Esto nos permite escribir condiciones cortas y limpias.Fíjate también en el caso final: ¿qué pasa si las pulsaciones medidas son **superiores a la FCM teórica**? Eso es físicamente posible (algunos atletas superan su FCM teórica) o puede ser un error de medición. Lo etiquetamos como caso a revisar para no dar una recomendación errónea.

```python
if porcentaje < 50:    zona = 'Z0 — Reposo / muy ligero'elif porcentaje < 60:    zona = 'Z1 — Calentamiento'elif porcentaje < 70:    zona = 'Z2 — Quema de grasa'elif porcentaje < 80:    zona = 'Z3 — Aeróbica'elif porcentaje < 90:    zona = 'Z4 — Umbral anaeróbico'elif porcentaje <= 100:    zona = 'Z5 — Máxima'else:    zona = '⚠️ Por encima de la FCM teórica (revisar)'print(f'Zona: {zona}')
```

## Paso 4: Probémoslo con varios casosVamos a comprobar que el programa clasifica correctamente. Usamos directamente valores en código para no tener que reintroducir la edad cada vez.

```python
def clasificar(edad, ppm):    fcm = 208 - 0.7 * edad    porcentaje = (ppm / fcm) * 100    if porcentaje < 50:    return 'Z0'    elif porcentaje < 60:  return 'Z1'    elif porcentaje < 70:  return 'Z2'    elif porcentaje < 80:  return 'Z3'    elif porcentaje < 90:  return 'Z4'    elif porcentaje <= 100: return 'Z5'    else:                  return '⚠️'casos = [(20, 95), (35, 130), (50, 165), (60, 175), (70, 195)]for edad, ppm in casos:    print(f'{edad} años, {ppm} ppm → {clasificar(edad, ppm)}')
```
