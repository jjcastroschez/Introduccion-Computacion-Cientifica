# 💊 Acumulación de fármacos — versión iterativa

## Recordatorio del [Tema 3](https://github.com/jjcastroschez/Introduccion-Computacion-Cientifica/blob/main/03_variables_tipos_simples/ejemplos/T3_Ejem_ICC.md)

En el Tema 3 resolvimos este problema usando la **fórmula cerrada** de la serie geométrica infinita:

$$ \text{nivel estacionario} = \frac{a}{1-r} $$

donde $a$ es la dosis diaria y $r = 1 - \dfrac{\text{porcentaje eliminado}}{100}$ es la fracción que se queda en el cuerpo al día siguiente.

Esa fórmula nos daba el **límite** al que tiende la concentración, pero **no nos decía nada del proceso**: ¿qué pasa el primer día? ¿Y el segundo? ¿Cuántos días tarda en estabilizarse?

## Lo que cambia con el Tema 4

Ahora sabemos usar bucles. Eso significa que podemos **simular el proceso día a día**, en lugar de saltar directamente al límite. Para cada día calculamos:

$$ \text{cantidad}_{\text{nuevo día}} = \text{cantidad}_{\text{día anterior}} \cdot r + a $$

Y vemos cómo evoluciona la concentración 🎯.

## Paso 1: Pedir los datos al usuario

Lo mismo que en el Tema 3, pero ahora añadimos un dato nuevo: **cuántos días queremos simular**.

```python
dosis = float(input('Dosis diaria (mg): '))
porcentaje_elim = float(input('Porcentaje eliminado cada día (0-100): '))
dias = int(input('Días que queremos simular: '))

r = 1 - porcentaje_elim / 100
print(f'Cada día queda en el cuerpo el {r*100:.0f}% de lo que había')
```

## Paso 2: Simular día a día con un bucle `for`

Empezamos con `cantidad = 0` (el paciente no tiene fármaco en el cuerpo) y aplicamos la **recurrencia** una vez por cada día:

```python
cantidad = 0

for dia in range(1, dias + 1):
    cantidad = cantidad * r + dosis
    print(f'Día {dia:2}: {cantidad:.4f} mg')
```

> 💡 **Observa lo que ha pasado**: en cada vuelta, `cantidad` se **acumula**. Es lo que llamamos un *acumulador*. Es uno de los patrones más comunes de programación con bucles.


> 📝 **¿Por qué un for?**: hemos usado un bucle `for` porque sabemos la cantidad de días exactos que queremos repetir el cálculo. Recuerda que el bucler `for` era el idóneo en estas situaciones. 

## Paso 3: Comparar con el resultado del Tema 3

El Tema 3 nos daba el valor *límite* (cuando los días tienden a infinito). Ahora podemos comprobar que, efectivamente, nuestra simulación **se acerca cada vez más** a ese valor:

```python
valor_limite = dosis / (1 - r)
diferencia = abs(cantidad - valor_limite)
print(f'Valor del Tema 3 (fórmula cerrada): {valor_limite:.4f} mg')
print(f'Valor del Tema 4 (simulación):     {cantidad:.4f} mg')
print(f'Diferencia tras {dias} días:        {diferencia:.4f} mg')
```

## Paso 4: ¿Cuándo paramos? Mejora con `while` y criterio de convergencia

El Tema 3 te decía cuál era el límite. El Tema 4 puede ir más allá: **¿cuántos días necesita el paciente para estar a menos de 0.01 mg del valor estacionario?**

Esto lo respondemos con un bucle `while` cuya condición de parada es que la **diferencia entre dos días consecutivos** sea menor que una tolerancia.

```python
TOLERANCIA = 0.01
MAX_DIAS = 1000    # red de seguridad para evitar el bucle infinito

cantidad_anterior = 0
cantidad = dosis      # tras el primer día
dia = 1

while abs(cantidad - cantidad_anterior) > TOLERANCIA and dia < MAX_DIAS:
    cantidad_anterior = cantidad
    cantidad = cantidad * r + dosis
    dia = dia + 1

print(f'En {dia} días el paciente alcanza el estado estacionario (precisión: {TOLERANCIA} mg).')
print(f'Cantidad final: {cantidad:.4f} mg')
```

## 🎯 Conceptos del Tema 4 que has practicado

* ✅ **Bucle `for in range`** para repetir un cálculo un número conocido de veces.
* ✅ **Variable acumuladora** (`cantidad`): es el patrón fundamental para sumar/acumular dentro de bucles.
* ✅ **Recurrencia matemática** implementada como reasignación dentro del bucle.
* ✅ **Bucle `while` con criterio de convergencia** y red de seguridad contra bucles infinitos.
