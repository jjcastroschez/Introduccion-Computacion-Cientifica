# 🔢 Raíz cuadrada por el método babilónico*(versión explicada paso a paso)*---## El problema¿Cómo calcula tu calculadora una raíz cuadrada? **No es magia**. Lo hace mediante un algoritmo iterativo: empieza con una estimación tosca y la va refinando hasta que la diferencia entre dos pasos consecutivos es despreciable.El algoritmo más conocido es el **método babilónico** (también llamado **método de Herón**). Lo usaban hace más de 2 000 años los matemáticos babilónicos para sacar raíces a mano. Tiene una propiedad maravillosa: en cada vuelta, **el número de cifras correctas se duplica**. Eso significa que en menos de 10 iteraciones tienes precisión de 15 decimales.## La fórmulaDada una estimación $x_n$ de $\sqrt{a}$, una mejor estimación es:$$x_{n+1} = \frac{1}{2}\left(x_n + \frac{a}{x_n}\right)$$La intuición es esta: si $x_n$ fuese exactamente $\sqrt{a}$, entonces $a/x_n$ también sería $\sqrt{a}$ y la media sería el propio valor. Si $x_n$ es **demasiado grande**, entonces $a/x_n$ es **demasiado pequeño**, y la media de ambos cae más cerca del valor real.

## ¿Por qué `while` y no `for`?Esta es la pregunta clave del ejemplo. **No sabemos cuántas iteraciones harán falta**: depende del valor de `a` y de la tolerancia que pidamos. Por eso usamos `while`: el bucle se repite **mientras** no hayamos convergido.Pero atención: **siempre que uses `while`, protégete contra los bucles infinitos**. Aunque la teoría dice que el método converge siempre, en la práctica un error de programación o un caso patológico (`a < 0`, por ejemplo) podría hacer que nunca se cumpliera la condición de salida. Por eso añadimos un **límite máximo de iteraciones**.

## Paso 1: Pedir el dato y definir parámetrosLos parámetros del algoritmo son la **tolerancia** (cuánto queremos afinar) y el **número máximo de iteraciones** (la red de seguridad).

```python
a = float(input('Calcular la raíz cuadrada de: '))TOLERANCIA = 1e-10    # Diferencia mínima entre iteraciones consecutivasMAX_ITER = 100        # Red de seguridad contra bucles infinitosprint(f'Calcularemos √{a} con precisión {TOLERANCIA}')
```

## Paso 2: Estimación inicialNecesitamos un valor de partida. Cuanto más cerca esté del verdadero $\sqrt{a}$, menos iteraciones harán falta. Una opción sencilla y robusta es empezar con $a/2$. (Existen estimaciones más sofisticadas basadas en la representación binaria del número, pero $a/2$ funciona perfectamente para este ejercicio.)

```python
x = a / 2print(f'Estimación inicial: x_0 = {x}')
```

## Paso 3: Bucle de convergenciaEl corazón del algoritmo. Cada vuelta calculamos `x_nuevo` con la fórmula. Si la diferencia con el `x` anterior es **menor que la tolerancia**, hemos terminado y salimos con `break`. Si no, actualizamos `x` y damos otra vuelta.

```python
iteracion = 0while iteracion < MAX_ITER:    x_nuevo = 0.5 * (x + a / x)    print(f'Iter {iteracion + 1}: x = {x_nuevo:.15f}')    if abs(x_nuevo - x) < TOLERANCIA:        x = x_nuevo        break       # ¡Hemos convergido!    x = x_nuevo    iteracion = iteracion + 1print(f'\nResultado: √{a} ≈ {x:.15f}')print(f'Iteraciones empleadas: {iteracion + 1}')
```

> 🎯 **Observa la salida**: con `a=2` solo hacen falta unas 5 iteraciones para tener precisión de 15 decimales. Increíble, ¿verdad? Compara con `math.sqrt`:

```python
import mathprint(f'Con math.sqrt: {math.sqrt(a):.15f}')print(f'Con nuestro método: {x:.15f}')print(f'Diferencia: {abs(math.sqrt(a) - x)}')
```

## ¿Y qué hay del límite máximo de iteraciones?Es nuestra **red de seguridad**. Si por alguna razón el algoritmo no convergiera, el bucle no sería infinito: pararía después de 100 iteraciones. Vamos a comprobar que se activa correctamente con un caso problemático:

```python
# Probemos con una tolerancia ridículamente pequeña, imposible de alcanzar# por la precisión finita de los floata = 2.0TOLERANCIA = 1e-50MAX_ITER = 20x = a / 2iteracion = 0convergencia = Falsewhile iteracion < MAX_ITER:    x_nuevo = 0.5 * (x + a / x)    if abs(x_nuevo - x) < TOLERANCIA:        convergencia = True        x = x_nuevo        break    x = x_nuevo    iteracion = iteracion + 1if convergencia:    print(f'Convergió en {iteracion + 1} iteraciones')else:    print(f'⚠️ No convergió en {MAX_ITER} iteraciones (límite alcanzado)')    print(f'Mejor aproximación: {x:.15f}')
```
