# 🛡️ Validación de entrada con `try-except`*(versión explicada paso a paso)*---## El problemaUn programa **profesional** nunca confía en que el usuario teclee lo correcto. Tiene que ser **robusto frente a errores**: si el usuario teclea letras donde se espera un número, o un número fuera de rango, el programa debe **avisar amablemente** y volver a pedir el dato. **Nunca** debe romperse.Vamos a programar una rutina que pide un número entero entre 1 y 100, y que sigue pidiéndolo hasta obtener una respuesta válida.

## Las dos cosas que pueden ir malDistingamos los **dos tipos de error** que puede cometer el usuario:1. **Error de tipo**: teclea algo que no es un número (`'hola'`, `'3.14'`, espacios en blanco...). Esto provoca que `int(entrada)` lance una **excepción** `ValueError`.2. **Error lógico**: teclea un número correcto, pero **fuera del rango** permitido (`-5`, `200`...). No hay excepción de Python: somos **nosotros** los que tenemos que decidir que ese valor no nos vale.Los dos casos requieren **mecanismos distintos**:* Para el error de tipo → `try-except`.* Para el error de rango → un `if` y un mensaje.## Paso 1: Estructura general del bucleNecesitamos un bucle que **se repita hasta que la entrada sea válida**. Como no sabemos cuántas veces el usuario se equivocará, usamos `while True` con `break` cuando todo esté bien:

````python
# Esqueleto:while True:    entrada = input('Introduce un número: ')    # ... validar ...    # si es válido: break    # si no: mensaje de error y continuar# (no ejecutamos esta celda; es solo el esquema)
````

> 💡 En el [Tema 4](../teoria/T4_ICC.md#-buenas-prácticas) se desaconseja `while True` cuando hay una condición clara, pero **este es uno de los casos legítimos** donde resulta natural: "haz esto repetidamente hasta que algo bueno ocurra". Es exactamente el patrón que en otros lenguajes se hace con `do-while`.

## Paso 2: Capturar el error de tipo con `try-except`El `try-except` envuelve la operación que **podría fallar**. Si falla, en lugar de romperse el programa, salta al `except` y ejecuta nuestra recuperación. El `continue` hace que volvamos al inicio del `while` para volver a pedir.

````python
while True:    entrada = input('Introduce un número entero: ')    try:        n = int(entrada)    except ValueError:        print(f'❌ "{entrada}" no es un número entero válido.')        continue           # vuelve al inicio del while    # Si llegamos aquí, n es un entero válido    print(f'Has tecleado el entero {n}')    break
````

> 🎯 **Prueba a teclear**: `42`, `hola`, `3.14`, `'  5  '` (con espacios), `''` (sólo Enter). Observa cómo cada caso se gestiona limpiamente, sin romper el programa.

## Paso 3: Validar también el rangoAhora añadimos la validación lógica: aunque el número sea entero, debe estar entre 1 y 100.

````python
while True:    entrada = input('Introduce un número entero entre 1 y 100: ')    try:        n = int(entrada)    except ValueError:        print(f'❌ "{entrada}" no es un número entero. Inténtalo de nuevo.')        continue    if 1 <= n <= 100:        break    print(f'❌ {n} está fuera del rango [1, 100]. Inténtalo de nuevo.')print(f'✅ Has introducido el número {n}. ¡Gracias!')
````

Es el equivalente a esconder la basura debajo de la alfombra.### ✅ Lo correctoSiempre **especifica el tipo** que esperas:```pythonexcept ValueError:    # solo capturamos esto, lo demás se propaga```### 💡 Mejora opcional: agrupar varios tiposSi hay varios tipos esperables, los puedes agrupar en una tupla:```pythonexcept (ValueError, TypeError):    print('Error de conversión')```
