## Recordatorio del [Tema 3](https://github.com/jjcastroschez/Introduccion-Computacion-Cientifica/blob/main/03_variables_tipos_simples/ejemplos/T3_Ejem_ICC.md)

En el Tema 3 calculamos el valor presente de una **renta perpetua** (un pago $a$ todos los años para siempre) usando la fórmula:

$$ VP_\infty = \frac{a}{i} $$

donde $i$ es la tasa de interés en tanto por uno. Pero la mayoría de productos financieros **no son perpetuos**: tienen una duración finita (un préstamo a 25 años, una pensión vitalicia, etc.).

## Lo que cambia con el Tema 4

El valor presente de una anualidad de $n$ años es:

$$ VP_n = \sum_{t=1}^{n} \frac{a}{(1+i)^t} $$

Antes no podíamos sumar $n$ términos cualesquiera (no teníamos bucles). Ahora sí 🎉.

## Paso 1: Pedir los datos

Lo mismo que en T3, pero añadiendo `annus`:

```python
pago_anual = float(input('Pago anual (€): '))
tasa_interes = float(input('Tasa de interés anual (en %): '))
annus = int(input('Años de la anualidad: '))

i = tasa_interes / 100
print(f'Vas a recibir {pago_anual} € al año durante {annus} años.')
```

## Paso 2: Acumular el valor presente con un `for`

Aquí vemos otra vez el patrón **variable acumuladora**. La inicializamos a 0 y le sumamos el valor descontado de cada flujo:

```python
valor_presente = 0

for t in range(1, annus + 1):
    descuento = pago_anual / ((1 + i) ** t)
    valor_presente = valor_presente + descuento
    print(f'Año {t:2}: aporta {descuento:8.2f} € → acumulado {valor_presente:8.2f} €')
```

## Paso 3: Comparación con el caso perpetuo del Tema 3

Comprueba que, al aumentar `annus`, el valor presente se **aproxima** al de la renta perpetua. Es decir, recuperamos el resultado del Tema 3 cuando $n$ es grande:

```python
perpetua = pago_anual / i
diferencia = perpetua - valor_presente
print(f'Anualidad de {annus} años: {valor_presente:.2f} €')
print(f'Renta perpetua (Tema 3):  {perpetua:.2f} €')
print(f'Diferencia: {diferencia:.2f} € (cuánto menos vale por ser finita)')
```

## Paso 4: Y al revés: ¿cuántos años hacen falta para que el VP supere X?

Aquí entra el `while`: como no sabemos cuántos años necesitamos, repetimos hasta cumplir la condición.

```python
objetivo = float(input('\nValor presente objetivo (€): '))
valor_presente = 0
t = 0

while valor_presente < objetivo:
    t = t + 1
    valor_presente = valor_presente + pago_anual / ((1 + i) ** t)
    if t > 200:    # red de seguridad
        print('⚠️ Objetivo inalcanzable con esta tasa y este pago.')
        break

if valor_presente >= objetivo:
    print(f'Se necesitan {t} años para superar los {objetivo} €.')
```

Como puedes ver hemos usado un `break` para salir del bucle, esto le puede quitar legibilidad al código aunque es habitual su uso entre la comunidad de programadores de Python, aquí lo hemos puesto para que veas un ejemplo de uso. 

Recuerda que la *existencia de un único punto de entrada - un único punto de salida* era un **principio fundamental e histórico de la programación estructurada**, formulado con rigor teórico, no obstante hoy en día lo heredamos y aplicamos más bien como un consejo de buenas prácticas flexible.

En tus programas, en la asignatura, deberás seguir los principios de la programación estructurada, por lo que tus bloques deberán tener un único punto de entrada y uno de salida. Ya tendrás tiempo de evolucionar y hacerlo de otra manera (que está bien que conozcas) 😜.  

```python
objetivo = float(input('\nValor presente objetivo (€): '))
valor_presente = 0
t = 0

while valor_presente < objetivo and t <= 200:
    t = t + 1
    valor_presente = valor_presente + pago_anual / ((1 + i) ** t)

if valor_presente >= objetivo:
    print(f'Se necesitan {t} años para superar los {objetivo} €.')
else:
    print('⚠️ Objetivo inalcanzable con esta tasa y este pago.')
```

> 📝 **Preferencia**: esta última es matemáticamente más elegante porque la condición del `while` te dice exactamente cuándo va a parar todo de un solo vistazo, aunque mezclamos cosas: lógica del programa (alcanzar el objetivo) con condición de seguridad (para evitar entrar en bucles infinitos). 

## 🎯 Conceptos del Tema 4 que has practicado

* ✅ **Bucle `for`** sumando flujo a flujo (variable acumuladora).
* ✅ **Bucle `while`** con condición de salida desconocida.
* ✅ **Convergencia**: la anualidad finita tiende a la renta perpetua del Tema 3 cuando $n \to \infty$.
* ✅ **`break`** combinado con red de seguridad para evitar bucles infinitos.
