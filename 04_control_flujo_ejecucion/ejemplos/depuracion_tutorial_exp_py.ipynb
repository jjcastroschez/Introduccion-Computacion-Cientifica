# 🐞 Tutorial de depuración

Un programa con bucles y condicionales que **no funciona como esperabas** rara vez se arregla "leyendo el código con cara de concentración" 😅. Lo que necesitas es **depurar**: una técnica sistemática para encontrar dónde está el error.

En este tutorial vas a aprender **tres técnicas de depuración**, ordenadas de menor a mayor potencia:

1. 🐛 **`print()` debugging** — la técnica universal y eterna.
2. 🔬 **Módulo `pdb`** — el depurador integrado de Python.
3. 💻 **Depurador visual de VS Code** — el más cómodo (lo veremos por capturas).

Como guía, partimos de un **programa con un bug deliberado** y lo arreglamos.

---

## El programa con bugs

Se supone que este programa calcula la suma de los **N primeros números pares** ($2 + 4 + 6 + \dots$). Pero algo va mal:

````python
n = 5      # queremos sumar los 5 primeros pares: 2+4+6+8+10 = 30

suma = 0
i = 0
while i < n:
    i = i + 1
    suma = suma + 2 * i
    i = i + 1                # ⚠️ ¿hay algo raro aquí?

print(f'Suma de los {n} primeros pares: {suma}')
````

Salida esperada: `Suma de los 5 primeros pares: 30`.  
Salida obtenida: `Suma de los 5 primeros pares: 18`. **¡Algo va mal!** 🐛

¿Lo ves a simple vista? Quizá sí, este ejemplo es muy sencillo. Pero imagina que el programa tiene 50 líneas, varios bucles anidados y muchas variables. **Necesitas técnica.**

---

## 🐛 Técnica 1: `print()` debugging

La técnica más vieja y más universal. Consiste en **insertar `print()` estratégicos** que te muestren *qué está pasando dentro del bucle*.

Las reglas de oro:

* Imprime **el valor de las variables** que crees que pueden estar mal.
* Imprime **dentro del bucle**, en cada vuelta, no solo al final.
* Marca tus mensajes con un prefijo distintivo (p.e. `'>>> DEBUG:'`) para distinguirlos de la salida normal.
* **Bórralos cuando hayas acabado** (no los dejes en el código de producción).

````python
n = 5

suma = 0
i = 0
while i < n:
    print(f'>>> DEBUG: entrando con i={i}, suma={suma}')
    i = i + 1
    suma = suma + 2 * i
    print(f'>>> DEBUG: tras la suma, i={i}, suma={suma}')
    i = i + 1
    print(f'>>> DEBUG: tras el segundo incremento, i={i}')

print(f'Suma: {suma}')
````

👀 ¿Lo ves ahora? Mira los valores de `i`: en cada vuelta, **`i` se incrementa dos veces** (una al principio y otra al final). En lugar de tomar los valores 1, 2, 3, 4, 5, salta a 2, 4, 6… y solo da **3 vueltas** antes de salir.

El bug está en la segunda línea `i = i + 1`: se nos coló por error.

### Versión corregida:

````python
n = 5

suma = 0
i = 0
while i < n:
    i = i + 1
    suma = suma + 2 * i      # ✅ ya solo incrementamos una vez

print(f'Suma de los {n} primeros pares: {suma}')
````

## 🔬 Técnica 2: el módulo `pdb`

`pdb` es el **depurador integrado de Python**. Te permite **pausar** la ejecución, **inspeccionar** las variables y **avanzar paso a paso** sin tener que añadir `print()` por todas partes.

La forma más sencilla de usarlo es con la función `breakpoint()` (incorporada desde Python 3.7). Cuando Python encuentra esa línea, **detiene el programa** y te abre una consola interactiva donde puedes escribir comandos.

### Comandos esenciales de `pdb`

| Comando | Forma corta | Qué hace |
| :--- | :---: | :--- |
| `next` | `n` | Ejecuta la siguiente línea |
| `continue` | `c` | Sigue hasta el siguiente breakpoint o el final |
| `print <var>` | `p <var>` | Muestra el valor de una variable |
| `list` | `l` | Muestra el código alrededor de la línea actual |
| `quit` | `q` | Sale del depurador |
| `help` | `h` | Muestra la ayuda |

> ⚠️ **Nota sobre los Notebooks**: `pdb` funciona perfectamente en scripts `.py`, pero en Jupyter es algo más limitado. Te recomiendo probar `pdb` en el archivo [`programa_con_bugs.py`](programa_con_bugs.py) desde una terminal:
>
> ```bash
> python programa_con_bugs.py
> ```
>
> El script ya tiene un `breakpoint()` insertado en la línea problemática para que practiques.

### Ejemplo de sesión `pdb`

Así se vería tu sesión al ejecutar `python programa_con_bugs.py`:

```text
> programa_con_bugs.py(11)<module>()
-> suma = suma + 2 * i
(Pdb) p i              # ¿Qué valor de i estamos procesando?
1
(Pdb) p suma           # ¿Cuál es la suma antes?
0
(Pdb) n                # Ejecuta la línea actual
> programa_con_bugs.py(12)<module>()
-> i = i + 1
(Pdb) p i              # 🚨 ¡incremento sospechoso!
1
(Pdb) n
(Pdb) p i              # ahora vale 2 sin haber procesado el siguiente par
2
(Pdb) c                # Continuar hasta el final
```

Verás claramente cómo `i` **avanza más rápido de lo esperado** en cada vuelta. Ahí está el bug.

## 💻 Técnica 3: el depurador visual de VS Code

VS Code (y prácticamente cualquier IDE moderno) tiene un depurador visual integrado. **Es la forma más cómoda y productiva** de depurar y la que te recomiendo aprender en serio.

### Cómo se usa, en 3 pasos

1. **Marca un *breakpoint***: haz click en el margen izquierdo de la línea donde quieres parar el programa. Aparecerá un punto rojo 🔴.
2. **Inicia la depuración**: pulsa **F5** (o ve a *Run → Start Debugging*). El programa se ejecutará y se detendrá en tu *breakpoint*.
3. **Avanza paso a paso** con los botones de la barra superior:
   * ▶️ *Continue* (F5): seguir hasta el siguiente *breakpoint*.
   * ⤵️ *Step Over* (F10): ejecutar la siguiente línea.
   * ⤴️ *Step Out* (Shift+F11): salir del bloque actual.

### Qué ves en la pantalla mientras depuras

* **Panel de variables** (izquierda): el valor actual de **todas las variables** locales y globales. Sin tener que poner `print()`.
* **Watch**: variables que quieres seguir de forma especial.
* **Consola de depuración**: donde puedes escribir expresiones Python que se evalúan en el contexto actual.

> 🎓 **Mi consejo**: dedica 30 minutos a [este tutorial oficial de VS Code para Python](https://code.visualstudio.com/docs/python/debugging). Es la inversión más rentable de todo el curso.

## 📋 ¿Qué técnica usar y cuándo?

| Situación | Técnica recomendada |
| :--- | :--- |
| Notebook de Jupyter | `print()` debugging |
| Script `.py` corto | `print()` o `breakpoint()` con `pdb` |
| Programa más complejo | Depurador de VS Code |
| En servidor remoto sin GUI | `pdb` en terminal |
| **Siempre que estés perdido** | Empieza con `print()`, escala si hace falta |
