# Simulación Farmacológica: El Estado Estacionario

Este cuaderno explica cómo la medicina se acumula en el cuerpo hasta alcanzar un equilibrio, utilizando el concepto de **series geométricas convergentes**.

## 1. El Problema
Un paciente toma una dosis diaria **$a$**. El cuerpo elimina un porcentaje cada día, por lo que sobrevive una fracción **$r$**.

La cantidad de medicamento en el cuerpo justo después de cada dosis es:
* **Día 1:** $a$
* **Día 2:** $a + a(r)$
* **Día 3:** $a + a(r) + a(r^2)$

Esto forma una serie geométrica: $S = a \sum_{n=0}^{k-1} r^n$

```python
# --- CONFIGURACIÓN DE LA SIMULACIÓN ---
dosis_diaria = 100  # mg
porcentaje_eliminado = 0.20
r = 1 - porcentaje_eliminado  # Fracción que permanece (0.8)

print(f"Dosis: {dosis_diaria} mg")
print(f"Factor de permanencia (r): {r}")
```

## 2. Cálculo de la Convergencia
A largo plazo ($k \to \infty$), la cantidad de medicamento no sube para siempre, sino que converge a:
$$S_\infty = \frac{a}{1 - r}$$

Calculémoslo con Python:

```python
limite_estacionario = dosis_diaria / (1 - r)

print(f"A largo plazo, el paciente tendrá un nivel estable de: {limite_estacionario} mg")
print(f"Este es el 'Estado Estacionario'.")
```

## 3. Simulación paso a paso
Vamos a ver cómo se acumula el fármaco durante los primeros 15 días:

```python
cantidad_en_cuerpo = 0
print("Día | Cantidad de Fármaco (mg)")
print("----------------------------")

for dia in range(1, 16):
    # Primero el cuerpo elimina lo del día anterior, luego tomamos la dosis
    cantidad_en_cuerpo = (cantidad_en_cuerpo * r) + dosis_diaria
    print(f"{dia:2d}  | {cantidad_en_cuerpo:.2f} mg")
```

### Conclusión Médica
Como se observa en los datos, después del día 10 la cantidad apenas cambia. Se ha estabilizado cerca de los **500 mg**. 

Si el nivel de toxicidad de este fármaco fuera de **600 mg**, el médico sabría que esta dosis de 100 mg es **segura** para el paciente.
