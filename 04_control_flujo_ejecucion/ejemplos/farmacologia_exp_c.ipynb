## ¿Por qué también en C?

El lenguaje **C** es uno de los lenguajes más usados en computación científica de alto rendimiento. Muchas bibliotecas científicas potentes que usarás en el futuro (BLAS, LAPACK, FFTW, GSL...) están escritas en C, y conocer el lenguaje te ayudará a leer su código fuente, entender mejor sus mensajes de error y, llegado el caso, optimizar partes críticas de tus programas.

Ver el **mismo algoritmo** implementado en C y en Python te permite además identificar qué cosas son **conceptuales** (comunes a todos los lenguajes) y cuáles son **sintácticas** (propias de cada uno).

## El código en C

```c
/*
 * Acumulación de fármacos — versión iterativa.
 * Tema 4 - Introducción a la Computación Científica (ICC).
 *
 * Compilar con: gcc -o farmacologia farmacologia_icc.c
 * Ejecutar con: ./farmacologia
 */

#include <stdio.h>

int main(void) {
    double dosis, porcentaje_elim, r, cantidad;
    int dias, dia;

    printf("Dosis diaria (mg): ");
    scanf("%lf", &dosis);
    printf("Porcentaje eliminado cada día (0-100): ");
    scanf("%lf", &porcentaje_elim);
    printf("Días que queremos simular: ");
    scanf("%d", &dias);

    r = 1.0 - porcentaje_elim / 100.0;
    cantidad = 0.0;

    for (dia = 1; dia <= dias; dia++) {
        cantidad = cantidad * r + dosis;
        printf("Día %3d: %.4f mg en el cuerpo\n", dia, cantidad);
    }

    printf("\nValor límite teórico (serie geométrica): %.4f mg\n",
           dosis / (1.0 - r));

    return 0;
}
```

## Diferencias clave con Python

Vamos a comentar **una por una** las diferencias entre la versión C y la Python:

### 1. Cabeceras y `main`

C exige declarar **al principio** las bibliotecas que vas a usar (`#include <stdio.h>` aporta las funciones `printf` y `scanf`) y empaquetar tu programa dentro de una función `main` que devuelve un entero (`return 0` significa "todo OK"). Python no necesita nada de esto: lo que escribas en un fichero se ejecuta tal cual.

### 2. Tipado estático

C es **tipado estático**: cada variable debe declararse con su tipo antes de usarla:

```c
double dosis, porcentaje_elim, r, cantidad;
int dias, dia;
```

El compilador te recordará si alguna vez intentas asignar un valor de un tipo incompatible. En Python esto no existe — las variables "agarran" el tipo del valor que les asignas.

### 3. Lectura del usuario: `scanf` vs `input`

En Python: `dosis = float(input("Dosis: "))`. En C la cosa es más compleja:

```c
printf("Dosis diaria (mg): ");
scanf("%lf", &dosis);
```

* `printf` solo imprime el mensaje (no lleva pregunta integrada).
* `scanf` lee la entrada, pero necesita un **especificador de formato** (`%lf` para `double`, `%d` para `int`) y la **dirección de memoria** de la variable (el `&` significa "dirección de").

### 4. El bucle `for`

C tiene el `for` clásico de tres partes (inicialización; condición; incremento):

```c
for (dia = 1; dia <= dias; dia++) {
    /* cuerpo del bucle */
}
```

Es **equivalente** al `for dia in range(1, dias + 1)` de Python. Fíjate que aquí los bloques **se delimitan con llaves** `{...}`, no con indentación.

### 5. Formato de salida

En Python usamos *f-strings*: `f"Día {dia}: {cantidad:.4f} mg"`. En C usamos especificadores dentro de `printf`:

```c
printf("Día %3d: %.4f mg en el cuerpo\n", dia, cantidad);
```

* `%3d` → un entero con al menos 3 espacios de ancho.
* `%.4f` → un real con 4 decimales.
* `\n` → salto de línea (en Python lo añade `print` automáticamente).

## 🎯 Lo que debes recordar

* ✅ El **algoritmo es idéntico** al de Python. Las diferencias son **sintácticas**, no conceptuales.
* ✅ C requiere **declarar el tipo** de cada variable y compilar antes de ejecutar.
* ✅ La **legibilidad** la dan los bloques con `{` y `}`, además de la indentación (que en C es solo convención, no obligatoria).
* ✅ Los **especificadores de formato** (`%d`, `%lf`, `%f`) controlan tanto la lectura como la escritura.
