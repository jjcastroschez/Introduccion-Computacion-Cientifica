# 🐞 Tutorial de depuraciónUn programa con bucles y condicionales que **no funciona como esperabas** rara vez se arregla "leyendo el código fijamente con cara de concentración" 😅. Lo que necesitas es **depurar**: una técnica sistemática para encontrar dónde está el error.En este tutorial vas a aprender **tres técnicas de depuración**, ordenadas de menor a mayor potencia:1. 🐛 **`print()` debugging** — la técnica universal y eterna.2. 🔬 **Módulo `pdb`** — el depurador integrado de Python.3. 💻 **Depurador visual de VS Code** — el más cómodo (lo veremos por capturas).Como guía, partimos de un **programa con un bug deliberado** y lo arreglamos.---## El programa con bugsSe supone que este programa cuenta cuántos **números pares** hay en una lista. Pero algo va mal:

````python
numeros = [4, 7, 12, 5, 8, 3, 10, 1, 6, 9]# Versión con bug — cuenta mal los parescontador = 0for n in numeros:    if n % 2 == 0:        contador = 1     # ⚠️ ¿Bug a propósito?print(f'Pares en la lista: {contador}')
````

Salida esperada: `Pares en la lista: 5` (porque hay 5 pares: 4, 12, 8, 10, 6).  Salida obtenida: `Pares en la lista: 1`. **¡Algo va mal!** 🐛¿Lo ves a simple vista? Quizá sí, este es muy fácil. Pero imagina que el programa tiene 50 líneas, varios bucles anidados y los datos vienen de un fichero. **Necesitas técnica.**

---## 🐛 Técnica 1: `print()` debuggingEs la técnica más vieja y más universal. Consiste en **insertar `print()` estratégicos** que te muestren *qué está pasando dentro del bucle*. Las reglas de oro:* Imprime **el valor de las variables** que crees que pueden estar mal.* Imprime **dentro del bucle**, en cada vuelta, no solo al final.* Marca tus mensajes con un prefijo distintivo (p.e. `'>>> DEBUG:'`) para distinguirlos de la salida normal.* **Bórralos cuando hayas acabado** (no los dejes en el código de producción).

````python
numeros = [4, 7, 12, 5, 8, 3, 10, 1, 6, 9]contador = 0for n in numeros:    print(f'>>> DEBUG: n={n}, contador antes={contador}')    if n % 2 == 0:        contador = 1        print(f'>>> DEBUG: ¡par! contador después={contador}')print(f'Pares en la lista: {contador}')
````

👀 ¿Lo ves ahora? Mira la línea `contador = 1`: en cada par, **estamos asignando 1**, no **incrementando** en 1. El bug es que escribimos `=` cuando queríamos `+=` (o `contador = contador + 1`).### Versión corregida

````python
numeros = [4, 7, 12, 5, 8, 3, 10, 1, 6, 9]contador = 0for n in numeros:    if n % 2 == 0:        contador = contador + 1     # ✅ ahora INCREMENTAprint(f'Pares en la lista: {contador}')
````

---## 🔬 Técnica 2: el módulo `pdb``pdb` es el **depurador integrado de Python**. Te permite **pausar** la ejecución, **inspeccionar** las variables y **avanzar paso a paso** sin tener que añadir `print()` por todas partes.La forma más sencilla de usarlo es con la función `breakpoint()` (incorporada desde Python 3.7). Cuando Python encuentra esa línea, **detiene el programa** y te abre una consola interactiva donde puedes escribir comandos.### Comandos esenciales de `pdb`| Comando | Forma corta | Qué hace || :--- | :---: | :--- || `next` | `n` | Ejecuta la siguiente línea || `step` | `s` | Entra dentro de una función llamada || `continue` | `c` | Sigue hasta el siguiente breakpoint || `print <var>` | `p <var>` | Muestra el valor de una variable || `list` | `l` | Muestra el código alrededor de la línea actual || `quit` | `q` | Sale del depurador || `help` | `h` | Muestra la ayuda |> ⚠️ **Nota sobre los Notebooks**: `pdb` funciona perfectamente en scripts `.py`, pero en Jupyter es algo más limitado. Te recomiendo probar `pdb` en el archivo [`programa_con_bugs.py`](programa_con_bugs.py) desde una terminal:>> ```bash> python programa_con_bugs.py> ```>> El script ya tiene un `breakpoint()` insertado en la línea problemática para que practiques.

### Ejemplo de sesión `pdb`Así se vería tu sesión al ejecutar `python programa_con_bugs.py`:```text> programa_con_bugs.py(7)<module>()-> contador = 1(Pdb) p n              # ¿Qué número estamos procesando?4(Pdb) p contador       # ¿Cuál es el contador antes?0(Pdb) n                # Ejecuta la línea actual> programa_con_bugs.py(8)<module>()-> for n in numeros:(Pdb) p contador       # ¿Y después?1(Pdb) c                # Continúa hasta el siguiente breakpoint o el final```Fíjate cómo `p contador` te confirma que **siempre vale 0 o 1**: el contador no crece, solo se queda en 1 cada vez que entra en el `if`. Ahí está el bug.

---## 💻 Técnica 3: el depurador visual de VS CodeVS Code (y prácticamente cualquier IDE moderno) tiene un depurador visual integrado. **Es la forma más cómoda y productiva** de depurar y la que te recomiendo aprender en serio.### Cómo se usa, en 3 pasos1. **Marca un *breakpoint***: haz click en el margen izquierdo de la línea donde quieres parar el programa. Aparecerá un punto rojo 🔴.2. **Inicia la depuración**: pulsa **F5** (o ve a *Run → Start Debugging*). El programa se ejecutará y se detendrá en tu *breakpoint*.3. **Avanza paso a paso** con los botones de la barra superior:   * ▶️ *Continue* (F5): seguir hasta el siguiente *breakpoint*.   * ⤵️ *Step Over* (F10): ejecutar la siguiente línea.   * ⤵️ *Step Into* (F11): entrar dentro de una función.   * ⤴️ *Step Out* (Shift+F11): salir de la función actual.### Qué ves en la pantalla mientras depuras* **Panel de variables** (izquierda): el valor actual de **todas las variables** locales y globales. Sin tener que poner `print()`.* **Pila de llamadas** (*Call Stack*): qué funciones están "abiertas" en este momento.* **Watch**: variables que quieres seguir de forma especial.* **Consola de depuración**: donde puedes escribir expresiones Python que se evalúan en el contexto actual.

> 🎓 **Mi consejo**: dedica 30 minutos a [este tutorial oficial de VS Code para Python](https://code.visualstudio.com/docs/python/debugging). Es la inversión más rentable de todo el curso.
