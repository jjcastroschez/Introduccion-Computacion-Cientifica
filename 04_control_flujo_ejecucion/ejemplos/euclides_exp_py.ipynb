## El problema

Dados dos enteros positivos $a$ y $b$, queremos calcular su **máximo común divisor**, es decir, el mayor entero que divide a ambos sin dejar resto.

El **algoritmo de Euclides** (siglo III a.C., uno de los algoritmos más antiguos que se conocen) se basa en una identidad muy elegante:

$$ mcd(a, b) = mcd(b, a \bmod b) $$

Es decir: el MCD no cambia si sustituimos el mayor de los dos por el **resto** de la división entre ellos. Como el resto siempre es menor que el divisor, los números se hacen cada vez más pequeños hasta que uno llega a 0. En ese momento, el otro es el MCD.

## ¿Por qué `while` y no `for`?

Antes de programar, **piensa**: ¿sabes de antemano cuántas vueltas hacen falta? **No.** Depende del valor concreto de $a$ y $b$. Por eso necesitamos `while`: repetiremos mientras `b != 0`.

## Ejemplo manual antes de programar

Vamos a calcular $mcd(48, 18)$ a mano:

| Vuelta | $a$ | $b$ | resto = a mod b |
| :---: | :---: | :---: | :---: |
| 1 | 48 | 18 | 48 mod 18 = 12 |
| 2 | 18 | 12 | 18 mod 12 = 6 |
| 3 | 12 | 6 | 12 mod 6 = 0 |
| 4 | 6 | 0 | (stop) |

Resultado: $mcd(48, 18) = 6$ ✅

Fíjate cómo en cada vuelta hacemos tres cosas: calcular el resto, mover `b` a `a`, mover el resto a `b`. **Eso es exactamente lo que pondremos en el bucle.**

## Paso 1: Pedir los datos

```python
a = int(input('Primer entero positivo (el mayor): '))
b = int(input('Segundo entero positivo (el menor): '))
print(f'Calculamos MCD({a}, {b})')
```

## Paso 2: Aplicar el bucle de Euclides

> 💡 **Truco mental**: imagina que estás haciendo la tabla anterior en papel. Cada vuelta del `while` rellena una fila.

```python
vuelta = 0
while b != 0:
    vuelta = vuelta + 1
    resto = a % b
    print(f'Vuelta {vuelta}: a={a:4}, b={b:4}, resto={resto}')
    a = b
    b = resto

print(f'\nMCD = {a}')
```

## Paso 3: Validar la entrada con `try-except`

El usuario podría teclear cualquier cosa. Vamos a proteger el programa para que no se rompa si la entrada no es válida:

```python
entrada_valida = False
while not entrada_valida:
    try:
        a = int(input('Primer entero positivo (el mayor): '))
        b = int(input('Segundo entero positivo (el menor): '))
    except ValueError:
        print('⚠️ Tienes que introducir un número entero.')
    else:
        if b > 0 and a >= b:
            entrada_valida = True
        else:
            print('⚠️ Tienes que introducir números enteros positivos, siendo el primero mayor o igual que el segundo.')   

# Guardamos los originales para mostrarlos al final
a_original = a
b_original = b

while b != 0:
    a, b = b, a % b   # asignación múltiple: muy elegante

print(f'MCD({a_original}, {b_original}) = {a}')
```

> 🎯 **Observa la línea `a, b = b, a % b`**: esto es **asignación múltiple**. Python evalúa primero el lado derecho `(b, a%b)` con los valores actuales, y luego asigna a `a` y `b` simultáneamente. Es equivalente a usar una variable auxiliar para `resto`, pero más limpio.

## 🎯 Conceptos del Tema 4 que has practicado

* ✅ **Bucle `while`** con condición de parada `b != 0`.
* ✅ **Reasignación de variables** dentro del bucle.
* ✅ **Manejo de excepciones** con `try-except` para validar entradas.
* ✅ **Variable centinela** con `entrada_valida` para chequear si la entrada es válida.
* ✅ **Asignación múltiple** `a, b = b, a % b`.
