# Zonas de entrenamiento por frecuencia cardíacaClasificación de las pulsaciones según las zonas estándar de entrenamiento, basadas en la Frecuencia Cardíaca Máxima (FCM) calculada con la fórmula de Tanaka:$$\text{FCM} = 208 - 0.7 \cdot \text{edad}$$

```python
edad = int(input('Edad: '))ppm = int(input('Pulsaciones actuales (ppm): '))fcm = 208 - 0.7 * edadporcentaje = (ppm / fcm) * 100if porcentaje < 50:    zona = 'Z0 — Reposo / muy ligero'elif porcentaje < 60:    zona = 'Z1 — Calentamiento'elif porcentaje < 70:    zona = 'Z2 — Quema de grasa'elif porcentaje < 80:    zona = 'Z3 — Aeróbica'elif porcentaje < 90:    zona = 'Z4 — Umbral anaeróbico'elif porcentaje <= 100:    zona = 'Z5 — Máxima'else:    zona = '⚠️ Por encima de la FCM teórica (revisar)'print(f'FCM estimada: {fcm:.0f} ppm')print(f'Trabajas al {porcentaje:.1f}% de tu FCM')print(f'Zona: {zona}')
```
