# 🔢 Algoritmo de Euclides — versión en C*(explicación del código equivalente al de Python)*---Veamos cómo se implementa el algoritmo de Euclides en C. Es uno de los algoritmos clásicos más bonitos: el código es tan corto en C como en Python, y nos sirve para ver cómo se hace la **validación de entrada** en C sin disponer del mecanismo `try-except`.

## El código en C```c/* * Algoritmo de Euclides para el cálculo del MCD de dos enteros positivos. * Tema 4 - Introducción a la Computación Científica (ICC). * * Compilar con: gcc -o euclides euclides_icc.c * Ejecutar con: ./euclides */#include <stdio.h>int main(void) {    int a, b, a_original, b_original, resto;    int entrada_valida = 0;    /* Bucle de validación de entrada equivalente al try-except de Python */    while (!entrada_valida) {        printf("Primer entero positivo: ");        if (scanf("%d", &a) != 1 || a <= 0) {            printf("  ⚠️ Tienes que introducir un entero positivo.\n");            while (getchar() != '\n');   /* vaciar buffer */            continue;        }        printf("Segundo entero positivo: ");        if (scanf("%d", &b) != 1 || b <= 0) {            printf("  ⚠️ Tienes que introducir un entero positivo.\n");            while (getchar() != '\n');            continue;        }        entrada_valida = 1;    }    a_original = a;    b_original = b;    while (b != 0) {        resto = a % b;        a = b;        b = resto;    }    printf("MCD(%d, %d) = %d\n", a_original, b_original, a);    return 0;}```

## El bucle de Euclides — comparaciónSi nos fijamos solo en el corazón del algoritmo, vemos lo cerca que están las dos versiones:**Python:**```pythonwhile b != 0:    a, b = b, a % b   # asignación múltiple```**C:**```cwhile (b != 0) {    resto = a % b;    a = b;    b = resto;}```C **no tiene asignación múltiple**, así que necesitamos la variable auxiliar `resto` que guarde el valor de `a % b` antes de pisar `b`. Es exactamente la traducción mental de lo que hace Python entre bastidores.
