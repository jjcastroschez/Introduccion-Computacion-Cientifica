# Raíz cuadrada por el método babilónicoCálculo iterativo de $\sqrt{a}$ usando la recurrencia:$$x_{n+1} = \frac{1}{2}\left(x_n + \frac{a}{x_n}\right)$$Sin usar `math.sqrt`.

```python
a = float(input('Calcular la raíz cuadrada de: '))TOLERANCIA = 1e-10MAX_ITER = 100x = a / 2     # estimación inicialiteracion = 0while iteracion < MAX_ITER:    x_nuevo = 0.5 * (x + a / x)    if abs(x_nuevo - x) < TOLERANCIA:        x = x_nuevo        break    x = x_nuevo    iteracion = iteracion + 1print(f'√{a} ≈ {x:.10f}')print(f'Iteraciones empleadas: {iteracion + 1}')
```
