# 🛡️ Validación robusta — versión en C*(comparación con el `try-except` de Python)*---El ejemplo de validación robusta es **el que más cambia** al pasar de Python a C, porque **C no tiene mecanismo de excepciones**. Veremos cómo se consigue el mismo objetivo con técnicas más manuales pero igualmente efectivas.

## El código en C```c/* * Validación robusta de entrada de usuario. * Tema 4 - Introducción a la Computación Científica (ICC). * * Compilar con: gcc -o validacion validacion_entrada_icc.c * * Nota: C NO dispone del manejo de excepciones (try-except) de Python. * En su lugar, hay que comprobar manualmente el valor de retorno de * scanf y vaciar el buffer de entrada cuando la lectura falla. */#include <stdio.h>int main(void) {    int n;    int valido = 0;    int caracter;    while (!valido) {        printf("Introduce un número entero entre 1 y 100: ");        if (scanf("%d", &n) != 1) {            /* scanf no consiguió leer un entero */            printf("  ❌ Eso no es un entero válido.\n");            /* Vaciamos el buffer hasta el siguiente salto de línea */            while ((caracter = getchar()) != '\n' && caracter != EOF);            continue;        }        if (n < 1 || n > 100) {            printf("  ❌ %d está fuera del rango [1, 100].\n", n);            continue;        }        valido = 1;    }    printf("✅ Has introducido el número %d. ¡Gracias!\n", n);    return 0;}```

## Comparación: `try-except` vs valor de retorno de `scanf`**Python (con excepciones):**```pythontry:    n = int(entrada)except ValueError:    print('No es un entero válido.')    continue```**C (sin excepciones, comprobando retorno):**```cif (scanf("%d", &n) != 1) {    printf("No es un entero válido.\n");    /* vaciar buffer */    while ((caracter = getchar()) != '\n' && caracter != EOF);    continue;}```### El truco del `getchar` y `EOF`* `getchar()` lee **un carácter** del buffer de entrada y lo devuelve.* `EOF` (*end of file*) es un valor especial que indica que ya no quedan caracteres por leer.* El bucle `while ((caracter = getchar()) != '\n' && caracter != EOF);` **consume y descarta** todo lo que haya en el buffer hasta encontrar un salto de línea (lo que el usuario teclea termina con un Enter, que se representa como `'\n'`) o el final del archivo.

## ¿Y la validación de rango?**Esto es idéntico en Python y en C**: una vez tienes el número correctamente leído, una sentencia `if` te dice si está fuera de rango:```cif (n < 1 || n > 100) {    printf("  ❌ %d está fuera del rango [1, 100].\n", n);    continue;}```Esto se parece mucho a Python:```pythonif n < 1 or n > 100:    print(f'  ❌ {n} está fuera del rango [1, 100].')    continue```Solo cambian:* `||` en C ↔ `or` en Python.* `printf("... %d ...\n", n)` en C ↔ f-strings en Python.
