## De acumuladores del Tema 4 a listas del Tema 6

## Paso 1: Leer notas hasta encontrar una línea en blanco

Simulamos la lectura del usuario con datos fijos para poder ejecutar la celda en el notebook:

```python
notas = []                               # empezamos con la lista vacía
print('Introduce las notas (línea vacía para terminar):')
entrada = input('Nota: ')
while entrada != '': # línea vacía: salimos
    try:
        nota = float(entrada)
    except ValueError:
        print(f"  ⚠️ '{entrada}' no es un número.")
    else:
        if 0 <= nota <= 10:
            notas.append(nota)               # 🔑 aquí crece la lista
        else:
            print(f"  ⚠️ Número '{entrada}' fuera del rango [0, 10].")
    finally:
        entrada = input('Nota: ')        

print(f'\nLista final: {notas}')
```

> 🔑 **La instrucción clave**: `notas.append(nota)`. Este método **modifica la lista in situ** añadiéndole un elemento al final. Es una operación característica de las listas (por ser **mutables**); las tuplas y cadenas no la tienen.

> ⚠️ **`append` devuelve `None`, no la lista**. Nunca escribas `notas = notas.append(nota)` — sobrescribirías `notas` con `None`.

## Paso 2: Media y otras estadísticas de una lista

Con la lista ya construida, calcular sus estadísticas es cuestión de una línea por cada una:

```python
def media(notas):
    return sum(notas) / len(notas)

print(f'Suma:    {sum(notas)}')
print(f'Notas introducidas: {len(notas)}')
print(f'Media:   {media(notas):.2f}')
print(f'Mínimo:  {min(notas)}')
print(f'Máximo:  {max(notas)}')
```

**Observa cómo estas funciones (`sum`, `len`, `min`, `max`) reciben la lista completa** y devuelven un solo valor. Son las operaciones más comunes sobre secuencias y funcionan igual sobre listas, tuplas y (algunas) cadenas.

## Paso 3: Desviación estándar

La desviación estándar requiere recorrer la lista **dos veces**: primero para calcular la media, luego para calcular la suma de las desviaciones al cuadrado. Se puede hacer con un bucle sencillo:

```python
import math

def desviacion_estandar(notas):
    mu = media(notas)
    suma_cuadrados = 0
    for x in notas:
        suma_cuadrados = suma_cuadrados + (x - mu) ** 2
    return math.sqrt(suma_cuadrados / len(notas))

print(f'Desv. estándar: {desviacion_estandar(notas):.4f}')
```

> 💡 **Fíjate en la ventaja de haber guardado las notas en una lista**: como la tenemos completa, podemos recorrerla **varias veces** sin problema. Si hubiéramos usado un acumulador como en el Tema 4, tendríamos que haber leído los datos dos veces (una para la suma, otra para las desviaciones), lo cual sería una lata.

## Paso 4: Función `contar_aprobados` con condición

Contar cuántos alumnos aprueban es un patrón clásico: recorremos la lista, incrementamos un contador si se cumple una condición:

```python
def contar_aprobados(notas, corte=5.0):
    aprobados = 0
    for nota in notas:
        if nota >= corte:
            aprobados = aprobados + 1
    return aprobados

print(f'Aprobados (≥5): {contar_aprobados(notas)}')
print(f'Notables (≥7): {contar_aprobados(notas, 7)}')
```

## Anticipo del estilo funcional

El mismo `contar_aprobados` se puede escribir con `filter` o con `sum`, aprovechando que en Python `True == 1` y `False == 0`:

```python
def contar_aprobados_v2(notas, corte=5.0):
    return sum(1 for n in notas if n >= corte)   # generador

def contar_aprobados_v3(notas, corte=5.0):
    return len(list(filter(lambda n: n >= corte, notas)))

print(contar_aprobados_v2(notas))
print(contar_aprobados_v3(notas))
```

Las tres versiones son correctas y devuelven el mismo resultado. La versión con `for/if` es la más didáctica; la de `sum + generador` es la más **pythónica**; la de `filter` es un ejemplo del estilo funcional que veremos con detalle en el ejemplo [funcional.ipynb](./funcional_exp_py.ipynb).
