# 📖 Registros en C: `struct`
*(comparación con la versión Python)*

---

En Python has visto tres formas de agrupar datos heterogéneos: `dict`, `namedtuple`, `dataclass`. En C, **solo hay una**: la palabra clave **`struct`**. Pero es la que dio origen a todas las demás, así que es importante conocerla.

## El código en C

```c
#include <stdio.h>

struct Libro {
    char titulo[80];
    char autor[60];
    int anio;
    int paginas;
};

void mostrar_libro(struct Libro *l) {
    printf("Título:  %s\n", l->titulo);
    printf("Autor:   %s\n", l->autor);
    printf("Año:     %d\n", l->anio);
    printf("Páginas: %d\n", l->paginas);
}

int main(void) {
    struct Libro l1 = {"Bowie: Una Biografía", "M. Hesse", 2019, 167};
    l1.paginas = 200;
    mostrar_libro(&l1);
    return 0;
}
```

El archivo completo está en [libros.c](./libros.c).

## Diferencias clave con Python

### 1. Declaración

En C hay que **declarar el tipo `struct`** antes de usarlo, indicando el tipo de cada campo:

```c
struct Libro {
    char titulo[80];   // cadena de hasta 80 caracteres
    char autor[60];
    int anio;
    int paginas;
};
```

En Python con `dataclass`:

```python
@dataclass
class Libro:
    titulo: str        # sin tamaño fijo
    autor: str
    anio: int
    paginas: int
```

**La gran diferencia**: en C las cadenas tienen un **tamaño máximo fijado en compilación**. Si defines `char titulo[80]` y luego intentas guardar un título de 100 caracteres, se produce un **desbordamiento de memoria** (¡y bugs graves!).

### 2. Crear una instancia

**C** (con inicialización directa por posición):
```c
struct Libro l1 = {"Bowie", "M. Hesse", 2019, 167};
```

**Python** (por nombre):
```python
l1 = Libro(titulo='Bowie', autor='M. Hesse', anio=2019, paginas=167)
```

### 3. Acceso a los campos

**C — con variable normal**: operador punto `.`
```c
l1.titulo
l1.anio
```

**C — con puntero**: operador flecha `->` (equivalente a `(*l).titulo`)
```c
l->titulo    // dentro de mostrar_libro(struct Libro *l)
```

**Python**: siempre punto
```python
l1.titulo
```

> 💡 **Sobre el operador flecha `->`**: aparece cuando trabajas con **punteros** a estructuras. En este curso no profundizaremos en punteros, solo tienes que reconocerlo cuando lo veas.

### 4. Paso a funciones

Pasar una `struct` completa a una función **por valor** en C haría una **copia completa** de la estructura (ineficiente si es grande). Por eso lo habitual es **pasar un puntero**:

```c
void mostrar_libro(struct Libro *l) {  // recibe un puntero
    printf("%s\n", l->titulo);       // acceso con ->
}

mostrar_libro(&l1);   // le pasamos la DIRECCIÓN de l1
```

En Python esto es transparente: los objetos siempre se pasan "por referencia de objeto", así que no hay que preocuparse. Es una de las simplificaciones más importantes que Python hace respecto a C.

### 5. Mutabilidad

En C, **todas las structs son mutables** por defecto:

```c
l1.paginas = 200;   // ✅ funciona sin más
```

En Python, depende de la opción elegida:
* `dict`: mutable ✅
* `namedtuple`: **inmutable** ❌
* `dataclass`: mutable ✅ (o inmutable con `@dataclass(frozen=True)`)

## 🎯 Lo que debes recordar

* ✅ El **concepto** de agrupar datos heterogéneos con nombres es **universal**. En C se llama `struct`, en Python `dict`/`namedtuple`/`dataclass`.
* ✅ En C, las **cadenas tienen tamaño máximo fijo**; en Python no.
* ✅ Para acceder a un campo desde un puntero, en C se usa `->` (equivalente a `(*p).campo`).
* ✅ Es habitual pasar structs a funciones **por puntero** para evitar copias innecesarias.
* ✅ En C, todas las structs son **mutables** por defecto.
