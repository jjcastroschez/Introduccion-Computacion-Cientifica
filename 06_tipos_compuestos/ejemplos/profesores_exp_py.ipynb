## Objetivo

Vamos a modelar un grupo de profesores y a hacer varias operaciones sobre ellos: recorrerlos, filtrar por curso, ordenar por criterio... Es un ejemplo típico donde se combinan **listas** y **tuplas**, y donde el **desempaquetado** hace el código mucho más legible.

## Modelado: lista de tuplas

¿Cómo representamos los datos? Para cada profesor tenemos varios campos (nombre, correo, asignatura, curso, cuatrimestre). Una **tupla** es perfecta para agrupar esos datos porque:

1. **No van a cambiar** una vez asignados a un profesor (son inmutables).
2. **El orden es fijo**: siempre pondremos primero el nombre, luego el correo, etc.

Y para tener **varios profesores**, los agrupamos en una **lista**:

```python
# Cada tupla: (nombre, correo, asignatura, curso, cuatrimestre)
profesores = [
    ('Pepe',        'pepe@uclm.es',     'ICC',         1, 2),
    ('Ricardo',     'ricardo@uclm.es',  'Álgebra',     1, 1),
    ('José Carlos', 'jcarlos@uclm.es',  'Análisis',    1, 1),
    ('Javier',      'javier@uclm.es',   'Geometría',   2, 1),
    ('Juan',        'juan@uclm.es',     'Estadística', 2, 2),
    ('Cristina',    'cris@uclm.es',     'Numérica',    3, 1),
]

print(f'Hay {len(profesores)} profesores en la lista.')
```

## Recorrido básico (sin desempaquetado)

Podemos recorrer la lista y acceder a los campos por índice:

```python
for prof in profesores:
    print(f'  {prof[0]:12} → {prof[2]}')
```

Funciona, pero **cuesta leer**: `prof[0]`, `prof[2]`... ¿qué era cada índice? Cada vez que veas el código tendrás que volver a comprobar qué había en cada posición. Es frágil: si cambias el orden de los campos, se rompe todo.

## Recorrido con desempaquetado (mucho mejor)

El **desempaquetado** te permite dar **nombres** a los elementos de la tupla directamente en el `for`:

```python
for nombre, correo, asig, curso, cuat in profesores:
    print(f'  {nombre:12} imparte {asig} en {curso}º curso, {cuat}º cuatrimestre')
```

**Fíjate en la diferencia de legibilidad**: usar `curso` en vez de `prof[3]` es infinitamente más claro. Y si mañana añades un nuevo campo a la tupla (por ejemplo, un despacho), solo tienes que actualizar la línea del `for` — el resto del código sigue funcionando.

> 🎯 **Regla de oro**: cuando iteras sobre una lista de tuplas o listas, **usa siempre el desempaquetado**. Es más legible y más robusto.

## Filtrar: solo los profesores de un curso

Con un `if` dentro del `for` seleccionamos los profesores de segundo:

```python
print('Profesores de 2º curso:')
for nombre, correo, asig, curso, cuat in profesores:
    if curso == 2:
        print(f'  {nombre} ({asig})')
```

## Ordenar por varios criterios

Aquí entra en juego una herramienta muy potente: `sorted()` con un parámetro `key` que **calcula un valor** por el que ordenar. Este `key` suele ser una **lambda** (Tema 5) que recibe cada elemento y devuelve por qué queremos ordenarlo:

```python
# Ordenar por curso (índice 3 de la tupla)
ordenados = sorted(profesores, key=lambda p: p[3])
for nombre, _, asig, curso, _ in ordenados:
    print(f'  {curso}º: {nombre} ({asig})')
```

> 💡 **Observa el uso de `_`** para variables cuyo valor no nos interesa (correo y cuatrimestre en este ejemplo). Es una convención Python: `_` significa "desempaquetar aquí, pero no me molestaré en darle un nombre porque no lo voy a usar".

## Ordenar por dos criterios a la vez

Si queremos ordenar **primero por curso y, en caso de empate, por cuatrimestre**, la lambda devuelve una **tupla**. Python ordena tuplas comparando elemento a elemento (Tema 6, slide 36):

```python
ordenados = sorted(profesores, key=lambda p: (p[3], p[4]))
for nombre, _, asig, curso, cuat in ordenados:
    print(f'  {curso}º-{cuat}º: {nombre} ({asig})')
```

**Genial**: primero salen los de primer curso-primer cuatrimestre, luego los de primero-segundo, luego los de segundo-primero, etc.
