## ¿Qué es un palíndromo?

Un **palíndromo** (del griego *palin dromein*, "que recorre a la inversa") es una palabra o frase que se lee igual de izquierda a derecha que de derecha a izquierda. Algunos ejemplos: *Ana*, *reconocer*, *Anita lava la tina*, *Somos o no somos*.

Es un problema clásico y muy elegante que ilustra tres características de las cadenas en Python:

1. **Métodos de string** (`lower`, `replace`).
2. **Slicing** con paso negativo (`cadena[::-1]`).
3. **Iteración** sobre los caracteres.

En este notebook lo resolveremos paso a paso, mostrando cómo se combina todo.

## Paso 1: la idea básica sin normalización

Vamos a empezar por lo más simple: dado un texto **sin espacios, signos ni mayúsculas raras**, comprobar si es un palíndromo consiste en comparar la cadena consigo misma **invertida**.

Y aquí es donde el **slicing con paso -1** brilla: `cadena[::-1]` devuelve la cadena invertida en una sola expresión.

```python
cadena = 'reconocer'
print(cadena[::-1])   # invertida
print(cadena == cadena[::-1])
```

**¡Espectacular!** Con una sola línea (`cadena == cadena[::-1]`) hemos resuelto el problema para casos simples.

## Paso 2: normalizar para casos reales

En la vida real, los palíndromos famosos suelen tener **espacios, signos de puntuación y mayúsculas**:

> *Anita lava la tina*

Si comparamos directamente, no funciona (`"Anita lava la tina" != "anit al aval atinA"`). Necesitamos **normalizar** el texto antes: pasarlo a minúsculas y quitar espacios y signos.

```python
def normalizar(texto):
    texto = texto.lower()
    texto = texto.replace(' ', '')
    for signo in ',.;:!?¡¿-\'\"()':
        texto = texto.replace(signo, '')
    return texto

print(normalizar('Anita lava la tina'))
```

> 💡 **Fíjate** en el uso repetido de `texto = texto.replace(...)`. Recuerda: las cadenas son **inmutables**. `.replace` no modifica la cadena, sino que **devuelve una nueva**. Por eso hay que reasignarla cada vez.

## Paso 3: combinar todo en la función `es_palindromo`

```python
def es_palindromo(texto):
    normalizado = texto.lower()
    normalizado = normalizado.replace(' ', '')
    for signo in ',.;:!?¡¿-\'\"()':
        normalizado = normalizado.replace(signo, '')
    return normalizado == normalizado[::-1]

print(es_palindromo('Ana'))                  # True
print(es_palindromo('hola'))                 # False
print(es_palindromo('Anita lava la tina'))   # True
print(es_palindromo('Somos o no somos'))     # True
print(es_palindromo('Casa'))                 # False
```

## ¿Cómo sería sin slicing?

Es instructivo ver **cómo se habría escrito en el Tema 4**, sin poder usar el slicing `[::-1]`:

```python
def es_palindromo_manual(texto):
    # Normalización (como antes)
    normalizado = texto.lower().replace(' ', '')
    for signo in ',.;:!?¡¿-\'\"()':
        normalizado = normalizado.replace(signo, '')
    
    # Comparar carácter por carácter desde los dos extremos hacia el centro
    i = 0
    j = len(normalizado) - 1
    while i < j:
        if normalizado[i] != normalizado[j]:
            return False
        i = i + 1
        j = j - 1
    return True

print(es_palindromo_manual('reconocer'))
print(es_palindromo_manual('hola'))
```

**Ambas versiones son correctas**, pero la primera (con slicing) es mucho **más concisa y legible**. Es un buen ejemplo de cómo los tipos de datos complejos permiten expresar ideas de forma más directa.
