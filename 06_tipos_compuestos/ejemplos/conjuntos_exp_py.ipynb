## Cuando el orden no importa y no quieres duplicados

Los **conjuntos** (`set`) son un tipo muy especial de secuencia con dos características que los hacen únicos:

1. **No permiten duplicados**. Si añades algo que ya está, no pasa nada.
2. **No están ordenados**. No puedes hacer `s[0]`, y el orden de aparición al iterar no está garantizado.

A cambio, ofrecen dos superpoderes:

* **Comprobación de pertenencia (`x in s`) ultrarrápida** incluso con millones de elementos.
* **Operaciones matemáticas de conjuntos** con sintaxis nativa (unión, intersección, diferencia...).

## Creación

Se crean con llaves `{}` o con la función `set()`:

```python
# Con llaves
a = {1, 2, 3, 4}
print(a, type(a))

# A partir de una lista con duplicados (¡se eliminan automáticamente!)
b = set([1, 2, 2, 3, 3, 3, 4])
print(b)
```

> ⚠️ **Cuidado**: `{}` **NO crea un conjunto vacío**, crea un **diccionario vacío**. Para un conjunto vacío debes usar `set()`:

```python
vacio_1 = {}         # ¡es un diccionario!
vacio_2 = set()      # este SÍ es un conjunto vacío
print(type(vacio_1))
print(type(vacio_2))
```

## Aplicación 1: eliminar duplicados

Un caso de uso muy común. Si tienes una lista con posibles repetidos y quieres **solo los elementos únicos**:

```python
notas_con_repes = [7, 5, 8, 5, 7, 9, 10, 7, 5, 8]
print(f'Notas totales:   {len(notas_con_repes)}')

notas_unicas = set(notas_con_repes)
print(f'Notas distintas: {len(notas_unicas)}')
print(f'Cuáles: {sorted(notas_unicas)}')
```

> 💡 **Fíjate en el `sorted(...)`**: como los conjuntos no tienen orden, si quieres mostrar sus elementos en un orden concreto, ordénalos antes con `sorted`. El resultado final será una lista (`list`) con los elementos ordenados.

## Aplicación 2: comprobar pertenencia rápidamente

Cuando compruebas `x in coleccion`, Python tiene que recorrerla. Con una **lista** tarda proporcionalmente al tamaño (lineal: O(n)). Con un **conjunto**, tarda tiempo casi constante (O(1)), aunque tenga millones de elementos.

```python
vocales = {'a', 'e', 'i', 'o', 'u'}

palabra = 'murcielago'
solo_vocales = ''
for letra in palabra:
    if letra in vocales:       # ← esta comprobación es súper rápida
        solo_vocales = solo_vocales + letra

print(f"Vocales de '{palabra}': '{solo_vocales}'")
```

## Aplicación 3: operaciones matemáticas

Este es el gran regalo de Python a los matemáticos. Los operadores `|`, `&`, `-`, `^` son **literalmente los mismos** que en teoría de conjuntos matemática ($\cup$, $\cap$, $\setminus$, $\triangle$):

```python
matriculados_algebra   = {'Ana', 'Luis', 'Marta', 'Pedro', 'Sofía'}
matriculados_analisis  = {'Luis', 'Marta', 'Carlos', 'Elena'}

print(f'Álgebra:  {matriculados_algebra}')
print(f'Análisis: {matriculados_analisis}')
```

```python
# Unión: estudiantes matriculados en al menos una asignatura
union = matriculados_algebra | matriculados_analisis
print(f'∪: {sorted(union)}')
```

```python
# Intersección: estudiantes matriculados en ambas
interseccion = matriculados_algebra & matriculados_analisis
print(f'∩: {sorted(interseccion)}')
```

```python
# Diferencia: solo en álgebra, no en análisis
solo_algebra = matriculados_algebra - matriculados_analisis
print(f'A \\ B: {sorted(solo_algebra)}')
```

```python
# Diferencia simétrica: solo en una de las dos, no en ambas
solo_una = matriculados_algebra ^ matriculados_analisis
print(f'△: {sorted(solo_una)}')
```

## Modificar un conjunto

Como los conjuntos son **mutables**, tienen métodos para añadir y quitar elementos:

```python
s = {1, 2, 3}
s.add(4)
print(f'Tras add(4): {s}')

s.add(2)                    # ya estaba: no pasa nada
print(f'Tras add(2): {s}')

s.discard(3)                # quita 3
print(f'Tras discard(3): {s}')

s.discard(99)               # 99 no estaba: no da error
print(f'Tras discard(99): {s}')

# .remove() también quita, pero da error si no existe
try:
    s.remove(99)
except KeyError as e:
    print(f'❌ remove(99) lanza KeyError: {e}')
```

> 🎯 **Regla práctica**: usa `.discard()` cuando te dé igual si el elemento estaba o no; usa `.remove()` cuando quieras asegurarte de que estaba (y aceptes el error si no).
