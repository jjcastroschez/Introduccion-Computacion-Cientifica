## ¿Por qué necesitamos más que listas y tuplas?

En el ejemplo de los profesores usamos una **tupla** para agrupar los datos de cada uno: `('Pepe', 'pepe@uclm.es', 'ICC', 1, 2)`. Funciona, pero tiene un problema: **hay que recordar en qué posición está cada campo**.

¿Era `p[3]` el curso o el cuatrimestre? ¿Y si cambio el orden? El código se vuelve frágil.

**Los diccionarios**, **namedtuples** y **dataclasses** resuelven este problema **dando nombres a los campos**. En este notebook los veremos los tres, para que sepas cuál elegir en cada situación.

## Opción 1: diccionario (`dict`)

El diccionario es la opción más flexible. Los campos son **pares clave-valor**:

```python
libro = {
    'titulo': 'Bowie: Una Biografía',
    'autor': 'M. Hesse',
    'anio': 2019,
    'paginas': 167,
}

print(libro['titulo'])
print(libro['anio'])
```

**Ventajas del diccionario**:
* Es **flexible**: puedes añadir campos nuevos en cualquier momento.
* Es **mutable**: puedes modificar los valores.
* Sintaxis muy legible.

**Añadir un campo nuevo** es trivial:

```python
libro['isbn'] = '978-84-264-0465-7'
print(libro)
```

### Método `.get()`: acceso seguro

Acceder con `libro['clave_inexistente']` lanza `KeyError`. Para acceso seguro con **valor por defecto**, usa `.get()`:

```python
# Con []: si no existe la clave, error
try:
    libro['editorial']
except KeyError as e:
    print(f'❌ KeyError: {e}')

# Con .get(): si no existe, devuelve None (o el default que le pases)
print(libro.get('editorial'))
print(libro.get('editorial', 'desconocida'))
```

### Iterar por un diccionario

Se puede recorrer de tres formas:

```python
print('Solo claves:')
for clave in libro:
    print(f'  {clave}')

print('\nSolo valores:')
for valor in libro.values():
    print(f'  {valor}')

print('\nParejas clave-valor:')
for clave, valor in libro.items():
    print(f'  {clave}: {valor}')
```

## Opción 2: `namedtuple` — tupla con nombres

Una `namedtuple` es una **tupla**, pero con **nombres para sus campos**. Se importa del módulo `collections`:

```python
from collections import namedtuple

# Definimos el 'molde'
Libro = namedtuple('Libro', ['titulo', 'autor', 'anio', 'paginas'])

# Creamos una instancia
l = Libro('Bowie: Una Biografía', 'M. Hesse', 2019, 167)

# Acceso por NOMBRE
print(l.titulo)
print(l.anio)

# Y también por ÍNDICE (porque es una tupla)
print(l[0])
```

**Ventajas de `namedtuple`**:
* Acceso **por nombre** (`l.titulo`) muy legible.
* También indexable como tupla (`l[0]`) y **desempaquetable**.
* Muy ligera en memoria.
* Es **inmutable**: no puedes modificar sus campos.

El desempaquetado funciona igual que con una tupla normal:

```python
titulo, autor, anio, paginas = l
print(f'{titulo}, {autor}, {anio}, {paginas}')
```

Y comprobamos que **es inmutable**:

```python
try:
    l.paginas = 200
except AttributeError as e:
    print(f'❌ AttributeError: {e}')
```

## Opción 3: `dataclass` — la opción moderna

Los `dataclass` fueron introducidos en Python 3.7. Son como pequeñas clases donde el compilador genera automáticamente el constructor, el `__repr__`, etc. Son **mutables** por defecto y muy cómodos:

```python
from dataclasses import dataclass

@dataclass
class LibroDC:
    titulo: str
    autor: str
    anio: int
    paginas: int

libro_dc = LibroDC('Bowie: Una Biografía', 'M. Hesse', 2019, 167)

print(libro_dc.titulo)
print(libro_dc)   # ¡Repr automático!
```

**Ventajas de `dataclass`**:
* Sintaxis muy limpia con type hints (como los que aprendiste en el [Tema 5](../../05_programacion_modular/README.md)).
* Es **mutable**.
* Genera automáticamente `__init__`, `__repr__`, `__eq__`, etc.
* Se integra con muchas librerías modernas (validadores, ORMs, etc.).

Y sí, es mutable:

```python
libro_dc.paginas = 200
print(libro_dc)
```

## 📊 Comparativa: ¿cuál elegir?

| Necesitas... | Usa |
|:---|:---|
| Un registro rápido, con campos que quizá cambian con el tiempo | **`dict`** |
| Una entidad **inmutable** (parámetros de una función, punto en el plano...) | **`namedtuple`** |
| Una entidad **mutable** con muchos campos | **`dataclass`** |
| Compatibilidad total (funciona en Python 2 y en muchos frameworks) | **`dict`** |

**Regla práctica para este curso**: usa **`dict`** cuando quieras experimentar y **`dataclass`** cuando estés modelando una entidad seria con varios campos fijos.

## 🎯 Conceptos del Tema 6 que has practicado

* ✅ **`namedtuple`**: definición con `namedtuple('Nombre', [...campos])`, acceso por nombre e índice.
* ✅ **`dataclass`**: decorador `@dataclass` con type hints.
