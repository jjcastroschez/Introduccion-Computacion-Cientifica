## Un ejemplo clásico donde el diccionario brilla

Uno de los usos más comunes de los diccionarios es **contar cuántas veces aparece cada elemento** en una colección. Vamos a hacerlo con palabras en un texto: dado un texto, queremos saber cuántas veces aparece cada palabra.

Este ejemplo integra:

* **Métodos de string**: `.lower()`, `.replace()`, `.split()`.
* **Diccionarios**: acceso, `.get()` con valor por defecto, `.items()`, `.values()`.
* **`sorted()` con `key`**: ordenar el resultado por frecuencia.

## Paso 1: normalizar y separar en palabras

Primero pasamos a minúsculas y quitamos signos de puntuación básicos:

````python
texto = 'La ciencia es una forma de pensar mucho más que un conjunto de conocimientos. La ciencia nos enseña a dudar.'

# Paso 1: normalización
texto_norm = texto.lower()
for signo in ',.;:!?¡¿()"\'':
    texto_norm = texto_norm.replace(signo, '')

print(texto_norm)
````

Ahora dividimos en palabras con `.split()`, que sin argumento separa por **espacios en blanco**:

````python
palabras = texto_norm.split()
print(palabras)
print(f'Total: {len(palabras)}')
````

## Paso 2: contar con un diccionario

El truco elegante es usar `dict.get(clave, valor_por_defecto)`. Si la palabra **no está** en el diccionario, `.get(palabra, 0)` devuelve `0` y le sumamos 1. Si **sí está**, devuelve su valor actual y también le sumamos 1. **Una sola línea** para cubrir los dos casos**:

````python
contador = {}
for palabra in palabras:
    contador[palabra] = contador.get(palabra, 0) + 1

print(contador)
````

> 💡 **Sin `.get()`, el código sería más verboso**:
>
> ```python
> if palabra in contador:
>     contador[palabra] = contador[palabra] + 1
> else:
>     contador[palabra] = 1
> ```
>
> ¡Cuatro líneas en vez de una! El patrón `dict.get(clave, 0) + 1` es tan común en Python que merece la pena memorizarlo.

## Paso 3: ordenar por frecuencia (de más a menos)

El método `.items()` devuelve un iterable de pares `(clave, valor)`. Lo pasamos a `sorted` con un `key` que extraiga el valor:

````python
ordenado = sorted(contador.items(), key=lambda par: par[1], reverse=True)

for palabra, veces in ordenado[:10]:      # las 10 más frecuentes
    marca = '★' * veces
    print(f'  {palabra:15} {veces}  {marca}')
````

## Paso 4: alternativa moderna con `collections.Counter`

Python tiene ya en su biblioteca estándar una **clase específica** para contar: `collections.Counter`. Hace lo mismo que hemos hecho a mano, pero en una línea:

````python
from collections import Counter

contador_c = Counter(palabras)
print(contador_c)

# Y tiene un método .most_common(n) que devuelve las N más frecuentes ya ordenadas
print(contador_c.most_common(5))
````

> 💡 **Cuándo usar cada uno**: para aprender los conceptos (`dict.get`, ordenar por valor), hazlo a mano. Para código de producción o si `Counter` te resulta cómodo, úsalo sin remordimientos. Ambas soluciones son perfectamente válidas.
